## 2.2 Bagging 思想 - 随机森林 Random Forest

#### 1. 为什么需要随机森林？
虽然 Bagging 可以缓解高方差模型的过拟合问题，<br>
但如果**基学习器之间高度相关（每棵树的分裂方式都差不多）**，那么 Bagging 的效果可能会大打折扣，因为它们可能会犯同样的错误，从而无法有效地降低方差。<br>
随机森林（Random Forest）通过引入特征随机选择的机制来增加基学习器之间的差异性，从而提高模型的性能。<br>

#### 2. 什么是随机森林？
随机森林（Random Forest）是一种基于 Bagging 思想的集成学习方法，通过在每棵树的训练过程中引入特征随机选择的机制来增加基学习器之间的差异性，从而提高模型的性能。<br>
随机森林的核心思想是通过随机选择特征来构建每棵树，从而增加树之间的差异性，减少模型的方差，提高整体性能。<br>
随机森林在每次分裂节点时，随机选择一部分特征来进行分裂，而不是使用所有的特征。这种随机选择特征的机制使得每棵树在训练过程中具有不同的结构和分裂方式，从而增加了基学习器之间的差异性，提高了模型的性能。<br>
随机森林在分类和回归问题中都表现出色，尤其适用于处理高维数据和具有复杂关系的数据集。<br>

#### 3. 随机森林的核心流程：<br>
假设有：
1. 1000个样本
2. 10个特征
3. 100颗树 <br>

核心流程如下：
1. 从1000个样本中有放回地抽取1000个样本，生成第一个训练集，样本数量仍然是1000个，但可能包含重复的样本。
2. 重复步骤1，生成100个不同的训练集，每个训练集都包含1000个样本，但由于是有放回的抽样，可能会包含重复的样本。
3. 在每个训练集上训练一个决策树模型。
  - 在训练过程中，在每次分裂节点时，不再直接针对10个特征进行分裂，而是随机选择m个特征进行分裂
  - m默认：对于分类问题，m = sqrt(10) ≈ 3；对于回归问题，m = 10/3 ≈ 3.33（通常取整数，即m=3）
  - 这种随机选择特征的机制使得每棵树在训练过程中具有不同的结构和分裂方式，从而增加了基学习器之间的差异性，提高了模型的性能。
4. 对于一个新的测试样本，使用所有的树进行预测，每棵树对每个类别进行投票，最终选择获得最多票数的类别作为最终的预测结果。
5. 使用OOB数据评估模型的性能。

#### 4. 为什么还需要随机选择特征？
1. 如果只做 BootStrap，那么某些强特征可能在每棵树的训练集中都被抽中，那么每棵树的分裂方式可能会非常相似，导致基学习器之间高度相关，从而无法有效地降低方差。
2. 强制特征选择本质上就是让每棵树在训练过程中具有不同的结构和分裂方式，从而增加了基学习器之间的差异性，降低相关性，提高了模型的性能。
