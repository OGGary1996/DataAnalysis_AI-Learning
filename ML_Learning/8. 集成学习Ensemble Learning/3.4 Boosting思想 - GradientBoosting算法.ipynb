## 3.4 Boosting思想 - GradientBoosting算法

#### 1. Gradient Boosting 是什么？
Gradient Boosting（梯度提升）是另外一种Boosting算法，与 AdaBoosting 不同，GradientBoosting 直接从损失函数出发，通过优化损失函数来训练弱学习器。<br>
Gradient Boosting 的核心思想是通过迭代地训练弱学习器，每次训练时关注之前模型的残差（即预测值与真实值之间的差异），从而逐步提升整体模型的准确性。<br>
比如：现在真实值为100：
1. 第一次模型预测值为 y_pred_1 = 70，此时残差 r_1 = 100 - 70 = 30；
2. 第二次模型的目标不再是直接尝试拟合100，而是尝试拟合上一次得到的残差 r_1 = 30;
   - 第二次的模型预测为 y_pred_2 = 25,此时新的残差 r_2 = 30 - 25 = 5;
3. 第三次模型的目标是拟合新的残差 r_2 = 5;
   - 第三次的模型预测为 y_pred_3 = 4，此时新的残差 r_3 = 5 - 4 = 1;
4. 最终的模型：
   - 最后的预测值为 = y_pred_1 + y_pred_2 + y_pred_3 = 70 + 25 + 4 = 99;
   - 最终的残差 r_final = 100 - 99= 1;

#### 2. Gradient Boosting 的核心思想：
1. 加法模型：Gradient Boosting 通过加法模型的方式将多个弱学习器组合成一个强学习器，每个弱学习器都试图拟合前一个模型的残差。
2. 梯度下降：Gradient Boosting 通过梯度下降的方式优化损失函数，在每一轮迭代中，计算当前模型的损失函数的负梯度（在回归问题中，就是残差)，并将损失函数最小化就转换为了在新的模型中的结果接近上一轮的残差，所以叫拟合残差。
  - 在回归中，损失函数通常是均方误差（MSE），负梯度就是残差；
  - 在分类中，损失函数通常是对数损失（Log Loss），负梯度则是样本的概率误差。

#### 3. Gradient Boosting 的核心流程：
1. 先设定目标，对于回归问题，目标函数（损失函数）是均方误差（MSE）；对于分类问题，目标函数是对数损失（Log Loss）。
2. **初始化模型**：选择一个简单的模型作为初始模型；通常是一个常数模型；比如对于回归问题，可以使用训练数据的平均值作为初始模型；对于分类问题，可以使用训练数据中类别的比例作为初始模型。
3. **迭代训练**：在每一轮迭代中：
   - 计算当前模型的损失函数的负梯度（在回归问题中，就是残差)
   - 将损失函数最小化就转换为了在新的模型中的结果接近上一轮的残差，所以叫拟合残差。
   - 更新模型：将新训练的弱学习器加入到现有模型中，并更新模型的预测结果。
4. **重复迭代**：继续迭代，直到满足停止条件（如达到预设的迭代次数或残差足够小）。
5. **输出最终模型**：最终的模型是所有弱学习器的加权组合，能够更好地捕捉数据的复杂模式，从而提高预测性能。
