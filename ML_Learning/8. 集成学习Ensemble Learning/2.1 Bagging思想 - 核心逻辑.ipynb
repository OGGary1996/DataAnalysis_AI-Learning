## 2.1 Bagging 思想 - 核心逻辑

#### 1. Bagging是什么？
Bagging（Bootstrap Aggregating）是一种集成学习方法，通过对训练数据进行有放回的抽样，生成多个不同的训练集，并在每个训练集上训练一个模型，最后通过投票或平均来得到最终的预测结果。Bagging 的核心思想是通过减少模型的方差来提高整体性能，从而获得更好的预测效果。

#### 2. Bagging 解决什么问题？
Bagging 主要解决模型的高方差问题，通过生成多个不同的训练集来训练多个模型，从而减少模型对训练数据的过拟合，提高模型的泛化能力。Bagging 适用于各种类型的模型，尤其是那些容易过拟合的模型，如决策树。

##### 2.1 高方差是什么？
高方差是指模型对训练数据的敏感程度较高，容易过拟合训练数据，导致在测试数据上的表现较差。
比如在决策树中，树的深度越大，模型的复杂度越高，容易过拟合训练数据；如果此时训练数据发生变化，模型的预测结果可能会发生较大的变化，表现出高方差的特征。
所以，这里的方差是指模型对训练数据的敏感程度，方差越高，模型越容易过拟合训练数据，导致在测试数据上的表现较差。

##### 2.2 Bagging 如何解决高方差问题？
Bagging 通过生成多个不同的训练集来训练多个高方差模型，最后让所有的模型进行 投票/平均 来得到最终的预测结果，从而减少模型对训练数据的过拟合，提高模型的泛化能力。<br>
关键词：**降方差**

#### 3. Bagging 的核心逻辑
Bagging 的核心逻辑可以总结为以下2步骤：
1. **BootStrap 数据抽样**：从原始训练数据集中通过有放回的抽样方法生成多个不同的训练集。每个训练集的大小通常与原始训练数据集相同，但由于是有放回的抽样，可能会包含重复的数据点。
2. **Aggregating 结果聚合**：训练好多个模型之后，通过投票（分类问题）或平均（回归问题）来聚合这些模型的预测结果，从而得到最终的预测结果。

##### 3.1 BootStrap 数据抽样
1. 从原始原始训练集中（大小为N）有放回的抽取N个样本，生成一个新的训练集。由于是有放回的抽样，可能会包含重复的数据点。
2. 重复上述步骤K次，生成K个不同的训练集（大小仍然为N）
3. 一个非常重要的结论：
  - 每次抽样后，平均有63.2%的数据被抽中，36.8%的数据没有被抽中（即所谓的 OOB 数据）。
  - 这是因为每个样本被抽中的概率为1 - (1 - 1/N)^N，当N趋近于无穷大时，这个概率趋近于1 - e^(-1) ≈ 0.632。
  - OOB（Out-Of-Bag）: 在每次抽样过程中，约有36.8%的数据没有被抽中，这些数据被称为 OOB 数据。OOB 数据可以用来评估模型的性能，因为它们没有被用于训练模型。

##### 3.2 Aggregating 结果聚合
1. 对于分类问题，使用投票法（Voting）来聚合模型的预测结果。每个模型对每个类别进行投票，最终选择获得最多票数的类别作为最终的预测结果。
2. 对于回归问题，使用平均法（Averaging）来聚合模型的预测结果。计算所有模型预测值的平均值作为最终的预测结果。

#### 4. Bagging 的完整流程示例：
假设有一个原始训练数据集包含N个样本，我们想要使用Bagging方法来训练一个分类模型。以下是Bagging的完整流程示例：
1. 从原始训练数据集中有放回地抽取N个样本，生成第一个训练集。
2. 在第一个训练集上训练一个分类模型，得到第一个模型。
3. 重复步骤1和步骤2，生成K个不同的训练集，并在每个训练集上训练一个分类模型，得到K个模型。
4. 对于一个新的测试样本，使用K个模型进行预测，每个模型对每个类别进行投票，最终选择获得最多票数的类别作为最终的预测结果。
5. 使用OOB数据评估模型的性能。对于每个训练集，使用没有被抽中的OOB数据来评估对应模型的性能，计算所有模型在OOB数据上的平均性能指标（如准确率、均方误差等）来评估整个Bagging模型的性能。

#### 5. Bagging 的重要前提 ⚠️
Bagging 的重要前提是基学习器（base learner）之间的差异性。只有当基学习器之间存在足够的差异性时，Bagging 才能有效地降低模型的方差，提高整体性能。<br>
如果基学习器之间高度相关，那么 Bagging 的效果可能会大打折扣，因为它们可能会犯同样的错误，从而无法有效地降低方差。<br>
这也是为什么在随机森林算法中，除了使用 Bagging 的思想外，还引入了特征随机选择的机制，以增加基学习器之间的差异性，从而提高模型的性能。

#### 6. OOB 数据评估

##### 6.1 OOB是什么
OOB（Out-Of-Bag）数据是指在 Bagging 过程中没有被抽中的数据。由于每个训练集是通过有放回的抽样生成的，约有36.8%的数据没有被抽中，这些数据被称为 OOB 数据。OOB 数据可以用来评估模型的性能，因为它们没有被用于训练模型。

##### 6.2 OOB 数据评估的流程：
1. 假设一共有100颗树，每颗树的训练集是通过有放回的抽样生成的，那么每颗树的训练集中约有63.2%的数据被抽中，36.8%的数据没有被抽中。
2. 对于每颗树都有它没有见过的数据（即 OOB 数据）
3. 用这些OOB数据分别对每颗树进行预测，得到每颗树在OOB数据上的预测结果。
4. 将每颗树在OOB数据上的预测结果进行聚合（投票或平均），得到整个Bagging模型在OOB数据上的预测结果。
5. 计算整个Bagging模型在OOB数据上的性能指标（如准确率、均方误差等），来评估模型的性能。
