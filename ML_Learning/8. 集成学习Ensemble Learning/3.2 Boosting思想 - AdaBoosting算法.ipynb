## 3.2 Boosting思想 - AdaBoosting


#### 1. AdaBoosting 是什么？
AdaBoosting（Adaptive Boosting）是一种Boosting算法，通过调整预测错误样本和预测正确样本来使得模型专注于错误分类的样本，从而提高模型的性能。<br>

#### 2. AdaBoosting 的核心思想
1. 串行训练：AdaBoosting 通过串行地训练多个弱学习器，每个弱学习器都试图纠正前一个弱学习器的错误。
2. 权重调整：在每一轮训练中，AdaBoosting 根据当前弱学习器的错误率调整样本权重，使得模型更加关注错误分类的样本。
3. 弱模型加权：最终的模型是所有弱学习器的加权组合，其中权重取决于每个弱学习器的性能，表现更好的弱学习器会获得更高的权重。
4. 最终加权投票：在预测阶段，AdaBoosting 通过加权投票的方式结合所有弱学习器的预测结果，输出最终的预测结果。

#### 3. AdaBoosting 的核心流程
1. **初始化权重**：为每个训练样本分配一个初始权重，通常是相等的；比如有100个样本，每个样本的初始权重为 1/100 = 0.01。
2. **训练弱学习器**：在每一轮迭代中，使用当前的样本权重训练一个弱学习器。
3. **计算错误率**：计算当前弱学习器的错误率，即错误分类的样本权重之和。
  - 错误率 error_rate = SUM(权重 * 预测结果指示函数)，
  - 其中预测结果指示函数为 1（如果预测错误）或 0（如果预测正确）；使得预测错误的样本错误率越高。
4. **计算弱学习器权重alpha**：根据错误率计算当前弱学习器的权重，通常使用以下公式：
  - alpha = 0.5 * ln((1 - error_rate) / error_rate)，
  - 其中 ln 是自然对数；如果错误率较低，alpha 会较大，表示该弱学习器更重要。
5. **更新样本权重**：根据当前弱学习器的预测结果调整样本权重：
  - 对于被正确分类的样本，权重会减少；对于被错误分类的样本，权重会增加。
  - 更新公式为：
    - w_i = w_i * exp(-alpha * y_i * h_i(x_i))，其中 w_i 是样本 i 的权重，y_i 是样本 i 的真实标签，h_i(x_i) 是弱学习器对样本 i 的预测结果。
6. **归一化权重**：将更新后的样本权重进行归一化，归一化公式为：
  - w_i = w_i / SUM(w_i)，确保所有样本权重的总和为1。
7. **重复迭代**：继续迭代，直到满足停止条件（如达到预设的迭代次数或错误率足够低）。
8. **输出最终模型**：最终的模型是所有弱学习器的加权组合，预测结果通过加权投票的方式输出。
