## 1.1 SVM 目标函数和约束函数推导 - 硬间隔 Hard Margin

#### 1. SVM 约束函数推导 <br>
SVM 的目标是找到一个最优分类超平面，将不同类别的样本点分开。为了确保分类的正确性，我们需要定义约束条件。<br>
对于线性可分的数据集，假设我们有一个训练集 {(x₁, y₁), (x₂, y₂), ..., (xₙ, yₙ)}，其中 xᵢ 是输入特征向量，yᵢ 是对应的类别标签（yᵢ ∈ {+1, -1}）。<br>
我们希望找到一个超平面，使得所有正类样本点都在超平面的一侧，所有负类样本点都在另一侧。<br>
这个超平面可以表示为：<br>
$$ w^T x + b = 0 $$<br>
其中，w 是法向量，b 是偏置项。<br>
为了确保分类的正确性，我们需要满足以下约束条件：<br>
- 对于正类样本点 (yᵢ = +1)：<br>
$$ w^T xᵢ + b ≥ 0 $$<br>
- 对于负类样本点 (yᵢ = -1)：<br>
$$ w^T xᵢ + b ≤ 0 $$<br><br>
由于这两个不等式需要同时满足，我们需要将它们结合起来，得到统一的约束条件：<br>
$$ yᵢ (w^T xᵢ + b) ≥ 0,  ∀ i = 1, 2, ..., n $$<br>
这个约束条件确保了所有样本点都被正确分类，并且距离超平面至少有一个单位的间隔。<br>
- 无论样本点属于正类还是负类，通过乘以其对应的标签 yᵢ，得到的结果为正，即表示判断正确。<br>

#### 2. 约束函数的标准化 <br>
为了简化计算，我们通常将约束条件进行标准化处理。<br>
我们可以将约束条件改写为：<br>
$$ yᵢ (w^T xᵢ + b) ≥ 1,  ∀ i = 1, 2, ..., n $$<br>
通过将右侧的 0 改为 1，我们实际上是在定义一个“间隔”，即样本点到超平面的距离至少为 1。<br>
这种标准化的约束条件是为了在计算Margin时，分子为1，从而简化了优化问题的求解。<br>
- 通过这种方式，我们确保了分类的正确性，同时也为最大化间隔提供了基础。<br>

#### 3. SVM 目标函数推导 <br>
在满足约束条件的前提下，我们希望最大化分类超平面与最近样本点之间的距离，这个距离被称为“间隔”（Margin）。<br>
间隔的计算公式为：<br>
$$ Margin = \frac{2}{||w||} $$<br>
为了最大化间隔，我们需要最小化 ||w||，即法向量的范数。<br>
为了简化计算，我们通常最小化 ||w||²/2。<br>
因此，SVM 的目标函数可以表示为：<br>
$$ \min \frac{1}{2} ||w||^2 $$<br>
结合约束条件，我们可以将 SVM 的优化问题表示为：<br>
$$ \min \frac{1}{2} ||w||^2 $$<br>
$$ s.t.  yᵢ (w^T xᵢ + b) ≥ 1,  ∀ i = 1, 2, ..., n $$<br>
这个优化问题的目标是找到一个最优的法向量 w 和偏置 b，使得间隔最大化，同时满足所有样本点的分类约束。<br>
- 通过求解这个优化问题，我们可以得到一个能够有效分类样本点的超平面。

#### 4. 硬间隔 SVM 的特点 <br>
硬间隔 SVM 适用于线性可分的数据集，即所有样本点都可以被一个超平面完全分开。<br>
在实际应用中，数据往往存在噪声和异常值，硬间隔 SVM 可能无法很好地处理这些情况。<br>
因此，在实际应用中，通常会使用软间隔 SVM 来处理线性不可分的数据集。<br>
- 软间隔 SVM 允许一定程度的分类错误，从而提高模型的鲁棒性。
