## 1.2 SVM 目标函数和约束函数推导 - 松弛变量&HingeLoss&C参数

#### 1. 引入松弛变量<br>
在实际应用中，数据集往往存在噪声和异常值，导致无法完全线性可分。为了处理这种情况，我们引入松弛变量 ξᵢ（xi_i），用于衡量样本点违反约束条件的程度。<br>
松弛变量 ξᵢ 定义为非负值（ξᵢ ≥ 0），表示样本点 i 到分类超平面的距离与间隔的差距。<br>
- 当 ξᵢ = 0 时，表示样本点被正确分类且位于间隔之外。<br>
- 当 0 < ξᵢ < 1 时，表示样本点被正确分类但位于间隔之内。<br>
- 当 ξᵢ ≥ 1 时，表示样本点被错误分类。<br>
引入松弛变量后，SVM 的约束条件可以表示为：<br>
$$ yᵢ (w^T xᵢ + b) ≥ 1 - ξᵢ,  ∀ i = 1, 2, ..., n $$<br>
这个约束条件允许样本点在一定程度上违反分类约束，从而提高模型的鲁棒性。

#### 2. 得到新的约束函数<br>
通过引入松弛变量 ξᵢ，我们得到了新的约束函数：<br>
$$ yᵢ (w^T xᵢ + b) ≥ 1 - ξᵢ,  ∀ i = 1, 2, ..., n $$<br>
这个约束函数允许样本点在一定程度上违反分类约束，从而提高模型的鲁棒性。<br>
- 松弛变量 ξᵢ 衡量了样本点违反约束条件的程度。<br>
- 通过调整松弛变量的值，我们可以控制模型对分类错误的容忍度。

#### 3. 新的目标函数<br>
为了同时最大化间隔和最小化分类错误，我们需要修改 SVM 的目标函数。<br>
新的目标函数可以表示为：<br>
$$ \min \frac{1}{2} ||w||^2 + C \sum_{i=1}^{n} ξᵢ $$<br>
其中，C 是一个正则化参数，用于平衡间隔最大化和分类错误最小化的权重。<br>
- 当 C 较大时，模型更倾向于最小化分类错误，可能导致间隔较小。<br>
- 当 C 较小时，模型更倾向于最大化间隔，可能允许更多的分类错误。<br>
通过调整参数 C，我们可以控制模型的复杂度和泛化能力。

#### 4. 引入 Hinge Loss<br>
为了量化分类错误的代价，我们引入了 Hinge Loss（铰链损失）函数。<br>
将新的约束函数化简后，我们可以得到 Hinge Loss 函数的定义。<br>
Hinge Loss 函数定义为：<br>
$$ L(yᵢ, f(xᵢ)) = \max(0, 1 - yᵢ f(xᵢ)) $$<br>
其中，f(xᵢ) = w^T xᵢ + b 是分类函数的输出。<br>
- 当样本点被正确分类且位于间隔之外时， Hinge Loss 为 0。<br>
- 当样本点被正确分类但位于间隔之内时， Hinge Loss 为正值，表示分类错误的代价。<br>
- 当样本点被错误分类时， Hinge Loss 也为正值，表示分类错误的代价。<br>
Hinge Loss 函数有效地量化了分类错误的代价，有助于优化 SVM 模型。

#### 5. 综合目标函数<br>
结合 Hinge Loss 函数，我们可以将 SVM 的目标函数重新表示为：<br>
$$ \min \frac{1}{2} ||w||^2 + C \sum_{i=1}^{n} L(yᵢ, f(xᵢ)) $$<br>
这个综合目标函数同时考虑了间隔最大化和分类错误的最小化。<br>
- 通过优化这个目标函数，我们可以得到一个能够有效分类样本点的超平面，同时具有良好的泛化能力。<br>
- 参数 C 的选择对于模型的性能至关重要，需要根据具体应用场景进行调整。
