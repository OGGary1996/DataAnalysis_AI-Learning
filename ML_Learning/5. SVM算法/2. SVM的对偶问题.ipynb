## 2. SVM 的对偶问题

#### 1. 引入拉格朗日乘子法<br>
为了求解 SVM 的优化问题，我们引入拉格朗日乘子法（Lagrange Multiplier Method）。<br>
拉格朗日乘子法是一种用于处理带约束优化问题的数学工具。<br>
通过引入拉格朗日乘子 αᵢ（alpha_i），我们可以将约束条件纳入目标函数中。<br>
对于 SVM 的优化问题，我们定义拉格朗日函数 L(w, b, α) 如下：<br>
$$ L(w, b, α) = \frac{1}{2} ||w||^2 - \sum_{i=1}^{n} αᵢ [yᵢ (w^T xᵢ + b) - 1] $$<br>
其中，αᵢ ≥ 0 是拉格朗日乘子，表示每个样本点对约束条件的贡献。<br>
- 通过引入拉格朗日乘子，我们将约束条件转化为目标函数的一部分，从而简化了优化问题的求解。

#### 2. 构建拉格朗日函数<br>
拉格朗日函数 L(w, b, α) 结合了 SVM 的目标函数和约束条件。<br>
通过构建拉格朗日函数，我们可以将原始的优化问题转化为无约束优化问题。<br>
拉格朗日函数的形式为：<br>
$$ L(w, b, α) = \frac{1}{2} ||w||^2 - \sum_{i=1}^{n} αᵢ [yᵢ (w^T xᵢ + b) - 1] $$<br>
- 这个函数包含了法向量 w、偏置 b 和拉格朗日乘子 αᵢ，反映了目标函数和约束条件之间的关系。

#### 3. 求解对偶问题<br>
为了求解 SVM 的对偶问题，我们需要对拉格朗日函数进行优化。<br>
首先，我们对 w 和 b 求偏导数，并将其设为零，以找到最优解。<br>
对 w 求偏导数：<br>
$$ \frac{∂L}{∂w} = w - \sum_{i=1}^{n} αᵢ yᵢ xᵢ = 0 $$<br>
解得：<br>
$$ w = \sum_{i=1}^{n} αᵢ yᵢ xᵢ $$<br>
对 b 求偏导数：<br>
$$ \frac{∂L}{∂b} = -\sum_{i=1}^{n} αᵢ yᵢ = 0 $$<br>
解得：<br>
$$ \sum_{i=1}^{n} αᵢ yᵢ = 0 $$<br>
将 w 的表达式代入拉格朗日函数，得到对偶问题的目标函数：<br>
$$ \max \sum_{i=1}^{n} αᵢ - \frac{1}{2} \sum_{i=1}^{n} \sum_{j=1}^{n} αᵢ αⱼ yᵢ yⱼ xᵢ^T xⱼ $$<br>
$$ s.t.  αᵢ ≥ 0,  ∀ i = 1, 2, ..., n $$<br>
$$ \sum_{i=1}^{n} αᵢ yᵢ = 0 $$<br>
- 通过求解这个对偶问题，我们可以得到最优的拉格朗日乘子 αᵢ，从而计算出最优的法向量 w 和偏置 b。

#### 4. 对偶问题的优点<br>
求解 SVM 的对偶问题具有多个优点：<br>
- 计算效率更高：对偶问题通常比原始问题更容易求解，尤其是在高维空间中。<br>
- 支持向量的识别：通过对偶问题的解，我们可以识别出支持向量，即那些对分类边界有贡献的样本点。<br>
- 核方法的应用：对偶问题的形式使得我们可以方便地引入核函数，从而处理非线性分类问题。<br>
- 更好的数值稳定性：对偶问题通常具有更好的数值稳定性，减少了计算中的误差积累。<br>
- 通过求解对偶问题，SVM 能够更有效地处理复杂的分类任务，提高模型的性能和泛化能力。

#### 5. 引入核函数<br>
在实际应用中，数据往往是非线性可分的。为了处理这种情况，我们引入核函数（Kernel Function）。<br>
核函数是一种将数据映射到高维空间的技术，使得在高维空间中数据变得线性可分。<br>
通过使用核函数，我们可以在不显式计算高维映射的情况下，计算样本点之间的内积。<br>
常用的核函数包括线性核、多项式核和径向基函数（RBF）核。<br>
- 通过引入核函数，SVM 能够处理复杂的非线性分类问题，扩展了其应用范围。
