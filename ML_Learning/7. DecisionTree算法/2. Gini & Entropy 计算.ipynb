## 2. Information Gain & Gini Decease 计算方法

决策树并不是直接使用特征来简单划分数据集，也不是通过简单的寻找最小的Gini和Entropy来划分数据集，而是通过计算**信息增益**和**基尼指数**来选择最优的特征进行划分。<br>

#### 1. Information Gain（信息增益）& Entropy（熵）
信息增益是决策树中常用的特征选择指标之一，它衡量了通过划分数据集所获得的信息增量。信息增益的计算基于熵的概念，熵是用来衡量数据集纯度的一种指标。<br>
信息增益的计算公式如下：
$$\text{Information Gain} = \text{Entropy}(S) - \sum_{i=1}^{n} \frac{|S_i|}{|S|} \text{Entropy}(S_i)$$
其中，$S$表示原始数据集，$S_i$表示通过某个特征划分后的子集，$|S|$表示原始数据集的样本数量，$|S_i|$表示子集的样本数量。<br>
熵的计算公式如下：
$$\text{Entropy}(S) = -\sum_{i=1}^{k} p_i \log_2(p_i)$$
其中，$k$表示数据集中不同类别的数量，$p_i$表示数据集中属于第$i$个类别的样本所占的比例。<br>
通过计算信息增益，决策树能够选择最优的特征进行划分，使得每个子集中的样本尽可能纯净，从而提高模型的性能和准确率。

#### 2. Gini Decrease & Gini Index（基尼指数）
Gini Decrease是决策树中另一种常用的特征选择指标之一， 它衡量了通过划分数据集所获得的基尼指数的减少量。Gini Decrease的计算基于基尼指数的概念，基尼指数是用来衡量数据集纯度的一种指标。<br>
Gini Decrease的计算公式如下：
$$\text{Gini Decrease} = \text{Gini Index}(S) - \sum_{i=1}^{n} \frac{|S_i|}{|S|} \text{Gini Index}(S_i)$$
其中，$S$表示原始数据集，$S_i$表示通过某个特征划分后的子集，$|S|$表示原始数据集的样本数量，$|S_i|$表示子集的样本数量。<br>
基尼指数是决策树中另一种常用的特征选择指标，它衡量了通过划分数据集所获得的纯度提升。基尼指数的计算公式如下：
$$\text{Gini Index} = 1 - \sum_{i=1}^{k} p_i^2$$
其中，$k$表示数据集中不同类别的数量，$p_i$表示数据集中属于第$i$个类别的样本所占的比例。<br>
基尼指数的计算方法与信息增益类似，通过计算基尼指数，决策树能够选择最优的特征进行划分，使得每个子集中的样本尽可能纯净，从而提高模型的性能和准确率。<br>
基尼指数的值越小，表示数据集的纯度越高，因此在决策树中，选择基尼指数最小的特征进行划分能够获得更好的分类效果。通过计算基尼指数，决策树能够逐步构建出一个树形结构，从而实现对数据的分类或回归。

#### 3. 选择最优特征进行划分
在决策树中，选择最优特征进行划分是构建树形结构的关键步骤。通过计算信息增益和基尼指数，决策树能够评估每个特征对于划分数据集的效果，从而选择最优的特征进行划分。<br>
选择最优特征的过程如下：
1. **计算信息增益**：对于每个特征，计算通过该特征划分数据集所获得的信息增益。选择信息增益最大的特征作为最优特征进行划分。
2. **计算基尼指数**：对于每个特征，计算通过该特征划分数据集所获得的基尼指数的减少量。选择基尼指数减少量最大的特征作为最优特征进行划分。
3. **比较信息增益和基尼指数**：在某些情况下，信息增益和基尼指数可能会给出不同的最优特征。此时，可以根据具体的应用场景和数据集的特点来选择使用哪种指标进行特征选择。
通过选择最优特征进行划分，决策树能够逐步构建出一个树形结构，从而实现对数据的分类或回归。选择最优特征能够提高模型的性能和准确率，使得每个子集中的样本尽可能纯净，从而提高模型的泛化能力。

#### 4. 总结
决策树并不是直接选用熵最低或者Gini最低的特征进行划分，而是通过计算 Information Gain 和 Gini Decrease 来选择最优的特征进行划分。<br>
通过计算信息增益和基尼指数，决策树能够评估每个特征对于划分数据集的效果，从而选择最优的特征进行划分。选择最优特征能够提高模型的性能和准确率，使得每个子集中的样本尽可能纯净，从而提高模型的泛化能力。
