## 1. 概率基本概念回顾

#### 1. 概率的定义
概率是一个数值，表示某个事件发生的可能性。它的取值范围在0到1之间，0表示事件不可能发生，1表示事件必然发生。概率可以通过频率来估计，即在大量试验中某个事件发生的次数与总试验次数的比值。

##### 1.1 概率的基本性质
- 非负性：对于任何事件A，概率P(A) ≥ 0。
- 规范化：对于样本空间S，概率P(S) = 1。
- 可加性：对于互斥事件A和B，概率P(A ∪ B) = P(A) + P(B)。

#### 2. 条件概率
条件概率是指在事件B已经发生的条件下，事件A发生的概率，记作P(A|B)。它可以通过以下公式计算：
P(A|B) = P(A ∩ B) / P(B)，其中P(A ∩ B)表示事件A和事件B同时发生的概率。

#### 3. 联合概率
联合概率是指两个或多个事件同时发生的概率，记作P(A ∩ B)。
计算公式为 P(A ∩ B) = P(A|B) * P(B)，其中P(A|B)表示在事件B发生的条件下事件A发生的概率，P(B)表示事件B的概率。
如果事件A和事件B是独立的，那么它们的联合概率可以通过以下公式计算：
P(A ∩ B) = P(A) * P(B)，其中P(A)和P(B)分别表示事件A和事件B的概率。

#### 4. 最大似然估计
最大似然估计是一种统计方法，用于估计模型参数，使得在给定数据集下，模型生成数据的概率最大。对于一个参数θ和一个数据集D，最大似然估计的目标是找到参数θ，使得P(D|θ)最大化。<br>
通常，我们通过对数似然函数来简化计算，即最大化log P(D|θ)。最大似然估计在机器学习中广泛应用于分类、回归等问题中，是一种重要的参数估计方法。

##### 4.1 第一步：定义似然函数
对于一个数据集D = {x1, x2, ..., xn}，**假设每个数据点xi都是独立同分布的**，那么似然函数(联合概率)可以表示为：<br>
L(θ) = P(D|θ) = Π P(xi|θ)，其中Π表示对所有数据点的乘积。

##### 4.2 第二步：对数似然函数
为了简化计算，我们通常对似然函数取对数，得到对数似然函数：<br>
log L(θ) = log P(D|θ) = Σ log P(xi|θ)，其中Σ表示对所有数据点的求和。

##### 4.3 第三步：最大化对数似然函数
最大似然估计的目标是找到参数θ，使得对数似然函数log L(θ)最大化。通常，我们通过求解对数似然函数的导数并设置为零来找到最优参数θ。<br>
d(log L(θ))/dθ = 0，解这个方程可以得到最大似然估计的参数值。
