## 2. 逻辑回归目标函数推导

#### 1 目标函数定义
逻辑回归是一种用于二分类问题的统计模型，其目标是通过最大化似然函数来估计模型参数。<br>
- 如果 y = 1，表示正类，那么 P(y=1|x) = hθ(x)，其中 hθ(x) 是逻辑回归模型的输出，表示输入 x 属于正类的概率。
- 如果 y = 0，表示负类，那么 P(y=0|x) = 1 - hθ(x)，表示输入 x 属于负类的概率。<br>
因此，单条样本的条件概率可以统一表示为：
P(y|x) = hθ(x)^y * (1 - hθ(x))^(1-y)，其中 y 是样本的真实标签，x 是样本的特征向量，hθ(x) 是模型的预测概率。<br>

#### 2. 似然函数定义
对于一个数据集 D = {(x1, y1), (x2, y2), ..., (xn, yn)}，假设每个样本都是独立同分布的，那么似然函数可以表示为：<br>
L(θ) = Π P(yi|xi) = Π [hθ(xi)^yi * (1 - hθ(xi))^(1-yi)]，其中 Π 表示对所有样本的乘积。

#### 3. 最大似然估计
为了简化计算，我们通常对似然函数取对数，得到对数似然函数：<br>
log L(θ) = log Π [hθ(xi)^yi * (1 - hθ(xi))^(1-yi)] = Σ [yi log hθ(xi) + (1 - yi) log (1 - hθ(xi))]，其中 Σ 表示对所有样本的求和。<br>
最大似然估计的目标是找到参数 θ，使得对数似然函数 log L(θ) 最大化。<br>

#### 4. 变成最小化问题
在实际优化过程中，我们通常将最大化对数似然函数转化为最小化负对数似然函数，即定义损失函数为：<br>
J(θ) = - log L(θ) = - Σ [yi log hθ(xi) + (1 - yi) log (1 - hθ(xi))]，其中 J(θ) 是逻辑回归的损失函数，也称为交叉熵损失函数。<br>

#### 5. 代入 hθ(xi) <br>
逻辑回归模型的输出 hθ(x) 通常定义为 sigmoid 函数，即 hθ(x) = 1 / (1 + exp(-θ^T x))，其中 θ 是模型的参数向量，x 是输入特征向量。<br>
将 hθ(xi) 代入损失函数 J(θ)，我们得到：
J(θ) = - Σ [yi log (1 / (1 + exp(-θ^T xi))) + (1 - yi) log (1 - 1 / (1 + exp(-θ^T xi)))]。<br>
进一步简化后，损失函数可以表示为：
J(θ) = - Σ [yi log (1 / (1 + exp(-θ^T xi))) + (1 - yi) log (exp(-θ^T xi) / (1 + exp(-θ^T xi)))]。<br>


#### 6. 加入正则化项
为了防止模型过拟合，我们通常在损失函数中加入正则化项。常见的正则化方法有 L1 正则化和 L2 正则化。<br>
- L1 正则化：R(θ) = λ Σ |θj|，其中 λ 是正则化强度，θj 是模型参数的第 j 个元素。
- L2 正则化：R(θ) = λ Σ θj^2。<br>
加入正则化项后的损失函数可以表示为：
J(θ) = - Σ [yi log hθ(xi) + (1 - yi) log (1 - hθ(xi))] + R(θ)，其中 R(θ) 是正则化项。<br>
通过最小化这个损失函数，我们可以得到逻辑回归模型的最优参数 θ，从而实现对二分类问题的预测。
