## 3.1 损失函数梯度下降 - 一元线性回归 <br>
梯度下降指的是：
1. 计算损失函数关于模型参数的梯度（偏导数），
2. 由于导数永远指向函数值上升最快的方向，如果导数为正，则表示原函数往正方向增加，若导数为负，则表示原函数往负方向增加。
3. 因此，我们如果想要梯度下降，则需要使用导数的负数来更新模型参数，从而使损失函数的值减小，如果导数为正，表示我们需要往负方向更新参数，反之亦然。
4. 通过不断迭代上述过程，最终我们可以找到损失函数的局部最小值，从而得到最优的模型参数。

梯度下降的公式：
$$ \theta = \theta - \alpha \cdot \nabla_{\theta} J(\theta) $$
其中，$\theta$表示模型参数，$\alpha$表示学习率，$\nabla_{\theta} J(\theta)$表示损失函数关于模型参数的梯度。 <br>
比如，对于一元线性回归模型，假设损失函数为平方误差损失函数，就是偏导数. <br>
$$ \sum{(y_i - (wx_i + b))^2} $$<br>
因此，我们可以通过不断迭代上述公式，来更新模型参数，从而使损失函数的值减小，最终得到最优的模型参数。

#### 手写模拟梯度下降过程，计算一元线性回归模型的参数 w 和 b。

##### 1. 准备数据

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 5, 4, 5], dtype=float)
```

##### 2. 初始化参数，以 w = 0, b = 0 开始， 学习率设置为0.01，迭代次数设置为10次

```python
w = 0.0 # 初始化权重
b = 0.0 # 初始化偏置
alpha = 0.01 # 学习率
epochs = 10 # 迭代次数
```

##### 3. 代入迭代公式，计算 w 和 b 的更新

```python
for epoch in range(epochs):
    # 计算ŷ
    y_pred = w * x + b

    # 计算误差
    error = y - y_pred

    # 计算损失 SSE
    loss = np.sum(error ** 2)

    # 计算梯度
    dw = -2 * np.sum(x * error)
    db = -2 * np.sum(error)

    # 更新参数
    w = w - alpha * dw
    b = b - alpha * db

    # 打印每次迭代的结果
    print(f"Epoch {epoch+1}: w={w:.4f}, b={b:.4f}, Loss={loss:.4f}")
```
