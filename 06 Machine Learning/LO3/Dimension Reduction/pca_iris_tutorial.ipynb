# PCA Tutorial Using the Iris Dataset
This notebook teaches students how PCA works, step-by-step, including standardization, covariance, eigenvalues, projection, and reconstruction.

## 1. Load and Standardize Data

```
import numpy as np
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler

# Load iris data
iris = load_iris()
X = iris.data  # (150, 4)

# Standardize
scaler = StandardScaler()
X_std = scaler.fit_transform(X)

X_std[:5]
```

## 2. Covariance Matrix

```
cov_mat = np.cov(X_std.T)
cov_mat
```

## 3. Eigenvalues and Eigenvectors

```
eig_vals, eig_vecs = np.linalg.eig(cov_mat)

print('Eigenvalues:', eig_vals)
print('\nEigenvectors:\n', eig_vecs)
```

## 4. Sort Eigenvalues

```
idx = np.argsort(eig_vals)[::-1]
eig_vals_sorted = eig_vals[idx]
eig_vecs_sorted = eig_vecs[:, idx]

eig_vals_sorted
```

## 5. PCA Projection (k=2)

```
k = 2
W = eig_vecs_sorted[:, :k]
X_pca = X_std @ W
X_pca[:5]
```

@ performs matrix multiplication according to linear algebra rules.

A @ B   ==   np.matmul(A, B)   ==   A.dot(B)

## 6. Reconstruction from 2 PCs

```
X_reconstructed = X_pca @ W.T
X_reconstructed_original = scaler.inverse_transform(X_reconstructed)

X_reconstructed_original[:5]
```

## 7. Compare Original and Reconstructed

```
print('Original data (first 5 rows):')
print(X[:5])

print('\nReconstructed data (first 5 rows):')
print(X_reconstructed_original[:5])
```
