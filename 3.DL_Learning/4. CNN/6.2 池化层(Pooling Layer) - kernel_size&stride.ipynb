## 6.2 池化层（Pooling Layer） - kernel_size&stride

#### 1. 池化层中的 kernel_size 是什么

##### 1.1 基本定义
在池化层中，kernel_size 表示的是：

池化窗口的大小

也就是说，池化层每次会取一个多大的局部区域来做池化计算。

例如：
* kernel_size = 2，通常表示窗口大小是 2 × 2
* kernel_size = 3，通常表示窗口大小是 3 × 3

所以，kernel_size 决定的是：

每次池化时，看多大一块区域。

##### 1.2 它和卷积层里的 kernel_size 类似，但作用不同
虽然名字一样都叫 kernel_size，

但在卷积层和池化层中，它们的意义并不完全一样。

在卷积层中：
* kernel_size 表示卷积核大小
* 卷积核内部有参数，会参与特征提取

在池化层中：
* kernel_size 表示池化窗口大小
* 池化窗口本身通常没有可学习参数
* 它只是决定“从多大的局部区域里选代表值”

#### 2. 池化层中的 stride 是什么

##### 2.1 基本定义
在池化层中，stride 表示的是：

池化窗口每次滑动时移动多少格。

也就是说，池化窗口在特征图上不是只计算一次，

而是会不断滑动。

而 stride 决定的，就是它每次往右、往下移动的步长。

例如：
* stride = 1：每次移动 1 格
* stride = 2：每次移动 2 格

##### 2.2 它的作用
池化层中的 stride 主要影响两件事：
* 输出特征图大小
* 池化区域之间是否重叠

所以 stride 越大：
* 输出越小
* 压缩越快

stride 越小：
* 输出越大
* 保留的信息越多

#### 3. 为什么池化层常见 kernel_size = 2, stride = 2

##### 3.1 最经典的搭配
在很多基础 CNN 中，最常见的池化层配置就是：

`nn.MaxPool2d(kernel_size=2, stride=2)`

或者：

`nn.AvgPool2d(kernel_size=2, stride=2)`

这表示：
* 每次看一个 2 × 2 区域
* 每次向右、向下各移动 2 格

##### 3.2 这样做的效果
这种设置会让池化窗口：
* 不重叠
* 刚好整齐地覆盖输入特征图

例如输入是：

`28 × 28`

经过 2 × 2、stride = 2 的池化后，输出通常就是：

`14 × 14`

也就是：

高和宽都缩小一半。

#### 4. 当 stride 小于 kernel_size 时会怎样

##### 4.1 会发生重叠池化
如果：
* kernel_size = 3
* stride = 2

那么池化窗口每次移动的距离，比窗口本身小。

这就意味着：

相邻池化区域之间会有重叠。

这种情况叫：

`Overlapping Pooling（重叠池化）`

##### 4.2 重叠池化的特点
它的特点是：
* 局部区域之间会共享一部分输入
* 输出会比 stride = kernel_size 时更大一些
* 信息保留更细一些

但同时也意味着：
* 计算量稍大
* 结构上不如标准 2×2, stride=2 那么简洁

#### 5. 池化层输出尺寸怎么受它们影响

##### 5.1 基本规律
池化层输出尺寸和卷积层类似，也会受到这些参数影响：
* 输入尺寸
* kernel_size
* stride
* padding

不过在基础池化中，最常见的是：
* padding = 0
* kernel_size = 2
* stride = 2

这时输出尺寸通常非常容易看：

>直接减半

##### 5.2 一个简单例子
假设输入特征图大小为：

`28 × 28`

使用：
* kernel_size = 2
* stride = 2

那么输出为：

`14 × 14`

如果输入是：

`32 × 32`

输出就是：

`16 × 16`

所以这种配置常常就是在做：

规则的 2 倍下采样
