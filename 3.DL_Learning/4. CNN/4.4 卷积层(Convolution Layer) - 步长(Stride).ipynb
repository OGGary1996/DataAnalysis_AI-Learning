## 4.4 卷积层（Convolution Layer） - 步长（Stride）

#### 1. Stride 是什么

##### 1.1 基本定义
它表示的是：

> 卷积核在输入图像上每次滑动时，移动多少格。

前面我们已经学习过：
* 卷积核会在输入图像上滑动
* 每滑动到一个位置，就进行一次卷积计算
* 最终得到特征图
而 Stride 决定的就是这个“滑动步伐”的大小。

##### 1.2 最直观的理解
你可以把卷积核想象成一个小窗口：
* 如果 stride = 1，表示每次移动 1 格
* 如果 stride = 2，表示每次移动 2 格
* 如果 stride = 3，表示每次移动 3 格

>所以步长越大，卷积核滑得越快，取样越稀疏。

>步长越小，卷积核滑得越细，扫描越密集。 🔍

#### 2. Stride 是如何工作的

##### 2.1 stride = 1
这是最常见的情况。

假设卷积核从左上角开始扫描：
* 第一次在第 1 列
* 第二次移动到第 2 列
* 第三次移动到第 3 列
* ……

也就是说，相邻两次卷积的位置是紧挨着的。

这种方式的特点是：
* 扫描最细致
* 输出信息更丰富
* 输出特征图通常更大

##### 2.2 stride = 2
如果步长改成 2，那么卷积核每次会跳过 1 格。

例如横向位置可能变成：
* 第 1 列
* 第 3 列
* 第 5 列
* ……

纵向也是同样的规律。

这种方式的特点是：
* 扫描更快
* 输出尺寸更小
* 相当于对输入做了更粗的采样

##### 2.3 一个直观比喻
可以把 stride 理解为你拿着放大镜看图：
* stride = 1：一格一格仔细看
* stride = 2：隔一格看一次
* stride = 3：跳得更远再看

所以：
>Stride 决定了卷积层“看图的密度”。

#### 3. Stride 的核心作用

##### 3.1 控制输出特征图大小
Stride 最直接的作用就是：
>控制输出尺寸。

步长越大，卷积核在输入上能停留的位置就越少，

因此最终输出特征图也会更小。

例如同样一个输入图像：
* stride = 1 → 输出更大
* stride = 2 → 输出更小

所以，除了 padding 之外，stride 也是影响输出尺寸的重要参数。

##### 3.2 控制信息保留的密度
步长小的时候：
* 卷积核扫描更细
* 保留的信息更多
* 特征提取更精细

步长大的时候：
* 卷积核扫描更粗
* 输出更紧凑
* 会丢失一部分细节

所以 stride 的本质也可以理解为：

>在“保留细节”和“压缩尺寸”之间做权衡。

##### 3.3 起到下采样的作用
当 stride > 1 时，卷积层不仅在提取特征，

还在一定程度上起到了下采样（downsampling）的作用。

例如：
* 输入较大
* 卷积后通过 stride = 2 直接让输出变小

这样可以减少后续计算量。

所以有些网络结构中，会通过卷积层本身的 stride 来完成尺寸压缩。

#### 4. Stride 对输出尺寸的影响

##### 4.1 输出尺寸公式
二维卷积中，单边输出尺寸公式为：

`输出尺寸 = (输入尺寸 + 2 × padding - 卷积核大小) / stride + 1`

如果结果不能整除，实际计算时通常取向下取整。

这个公式里，stride 在分母位置，所以很明显：

>stride 越大，输出尺寸越小。

##### 4.2 stride=1示例
假设：
* 输入尺寸：28
* 卷积核大小：3
* padding = 0
* stride = 1

那么输出尺寸：

`(28 - 3) / 1 + 1 = 26`

输出为：

`26 × 26`

##### 4.3 改成 stride = 2
如果其他条件不变，只把 stride 改成 2：
* 输入尺寸：28
* 卷积核大小：3
* padding = 0
* stride = 2

那么输出尺寸：

`(28 - 3) / 2 + 1 = 13.5`

实际要取整，所以输出为：

`13 × 13`

可以看出，步长从 1 变成 2 后，输出尺寸明显缩小了。

##### 4.4 实际意义
对于实际任务来说：
* stride = 1 更适合前期细致提取笔画特征
* stride = 2 更适合在后续适当压缩尺寸、减少计算量

>所以在实际 CNN 中，浅层常常更偏向保留细节，

>而后续层才逐步压缩空间尺寸。

#### 5. Stride 和 Padding 的关系

##### 5.1 两者都会影响输出尺寸
我们刚学过 padding，现在又学 stride。

这两个参数都会影响卷积输出大小：
* padding：决定是否给边缘“留空间”
* stride：决定卷积核滑动得多快

所以在计算输出尺寸时，它们总是一起出现。

##### 5.2 一个常见组合
例如：
* kernel_size = 3
* padding = 1
* stride = 1

这是最经典的组合之一，特点是：

>提取特征，同时保持尺寸不变

而如果改成：
* kernel_size = 3
* padding = 1
* stride = 2

则常常表示：

>提取特征的同时进行下采样

#### 6. 实际使用中的经验

##### 6.1 最常见的是 stride = 1
在很多基础卷积层中，最常见的选择就是：

`stride = 1`

因为这样更适合细致提取特征。

##### 6.2 压缩尺寸时常用 stride = 2
如果想让特征图变小，常见做法有两种：
* 用池化层压缩
* 用 stride = 2 的卷积层压缩

所以 stride = 2 也非常常见。

##### 6.3 前期不要随便把 stride 设太大
因为步长过大意味着跳得太快，

可能很多细节还没来得及学到就被跳过去了。

所以初学阶段可以先记住：
* 浅层：更常见 stride = 1
* 压缩：常见 stride = 2
