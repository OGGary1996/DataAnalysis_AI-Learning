## 5. 激活层激活函数

#### 1. 为什么卷积层后面还需要激活函数

##### 1.1 卷积层本身做了什么
我们前面已经学习过，卷积层的核心作用是：
* 用卷积核提取局部特征
* 得到一张或多张特征图

但要注意，卷积运算本质上仍然是线性运算。

因为它做的还是：
* 乘法
* 加法
* 再加偏置

也就是说，卷积层虽然结构上比全连接层更适合图像，

但从数学本质来看，它和线性层一样，仍然属于线性变换。

##### 1.2 如果没有激活函数会怎样
如果卷积层后面不加激活函数，那么即使堆很多层卷积层，

整体上仍然可以看成是一个更复杂一点的线性变换。

这就会带来一个问题：

>模型的表达能力会非常受限。

因为现实中的图像特征、类别边界、模式关系，大多数都不是简单的线性关系。

所以如果没有激活函数，CNN 很难学到复杂的非线性模式。

##### 1.3 激活函数的核心作用
激活函数最核心的作用就是：

给卷积层加入非线性能力。

有了激活函数之后，网络就不再只是简单地“线性提取特征”，

而是可以逐层学习越来越复杂的模式，例如：
* 边缘到纹理
* 纹理到局部形状
* 局部形状到整体目标

所以你可以把它理解为：

>卷积层负责提特征，激活函数负责让这些特征变得“可组合、可表达复杂关系”。

#### 2. 激活函数如何作用

##### 2.1 一般放在卷积层后面
在 CNN 中，最常见的顺序是：

`卷积层 → 激活函数`

也就是说：
* 卷积层先输出线性结果
* 激活函数再对这些结果逐元素处理
* 得到激活后的特征图

这和我们之前学过的 MLP 是一致的。

##### 2.2 一个常见结构
基础 CNN 中，经常能看到这样的结构：

`Conv → ReLU → Pool`

或者：

`Conv → BatchNorm → ReLU`

激活函数不是单独存在的，它通常紧跟在卷积层后面。

##### 2.3 它是如何作用在特征图上的
卷积层输出的结果是一张特征图，或者多通道特征图。

激活函数会对这些特征图中的每一个元素逐个处理。

例如：

卷积输出某一小块区域为：
```
-2   1   3
 0  -1   4
 2  -3   5
```
如果后面接 ReLU，那么激活后会变成：
```
0   1   3
0   0   4
2   0   5
```
也就是说：

激活函数是逐元素作用的，不会改变特征图的形状。

##### 2.4 结合手写数字识别来理解
（1）卷积后的特征图

假设输入是一张 `28 × 28` 的手写数字图像，

经过第一层卷积后，得到一个特征图中的某一部分：
```
-1.2   0.5   2.1
 1.8  -0.7   0.3
-2.0   1.1   3.4
```
这些值表示：
* 某个卷积核在不同位置上的响应强弱

---

（2）经过 ReLU 后
如果接一个 ReLU，那么输出变成：
```
0     0.5   2.1
1.8   0     0.3
0     1.1   3.4
```
这说明：
* 响应为负的位置被清零
* 响应为正的位置被保留


#### 3. 为什么卷积层特别需要激活函数

##### 3.1 图像特征本身是复杂的
图像中的模式不是简单的线性组合就能表达清楚的。

例如：
* 一个数字“8”和“3”的区别
* 一张猫脸和狗脸的区别
* 一块病灶和正常区域的区别

这些都不是“像素做一次线性加权”就能很好分开的。

所以 CNN 必须借助激活函数，让网络学会更复杂的判断方式。

##### 3.2 多层卷积的意义要靠非线性体现
我们总说 CNN 可以逐层提取特征：
* 浅层提取边缘
* 中层提取纹理和局部形状
* 深层提取高级语义特征

但这个“逐层越来越复杂”的前提就是：

层与层之间必须有非线性变换。

如果没有激活函数，那么多层卷积叠起来的意义就会大打折扣。

##### 3.3 激活函数帮助特征筛选
除了加入非线性，激活函数在某种程度上也起到了筛选响应值的作用。

例如 ReLU 会把负值压成 0，这相当于：
* 把不重要或负向响应去掉
* 保留更强的正向激活

这会让特征表达更加稀疏、更加突出重点。

#### 4. 卷积层后使用 ReLU 的实际意义

##### 4.1 让特征图更有选择性
卷积层输出后，不是所有位置都同样重要。

经过 ReLU 后：
* 强响应位置被保留
* 弱或负响应位置被抑制

这样特征图会更突出真正有用的区域。

##### 4.2 形成稀疏激活
很多值被 ReLU 压成 0，

这就会让特征图出现“很多地方不激活，少数地方激活明显”的现象。

这叫做：

>稀疏激活（Sparse Activation）

这种效果往往是有利的，因为它能让模型更聚焦于关键特征。

##### 4.3 有助于后续层继续提取模式
前一层经过激活函数之后，输出的特征更清晰，

后面的卷积层就更容易基于这些特征继续组合出更复杂的模式。

所以可以理解为：

>激活函数像是在帮卷积层做一次“特征清洗和突出重点”。

#### 5. 卷积层中最常见的激活函数：ReLU

##### 5.1 ReLU 的定义
在 CNN 中，最常用的激活函数就是：

>ReLU（Rectified Linear Unit）

它的公式非常简单：

`f(x) = max(0, x)`

意思就是：
* 如果 x > 0，输出 x
* 如果 x ≤ 0，输出 0

##### 5.2 为什么 ReLU 特别适合 CNN
ReLU 在卷积神经网络中非常常见，主要有几个原因：
* 计算简单
* 收敛速度通常更快
* 能缓解梯度消失问题
* 会让输出更稀疏

所以在大多数基础 CNN 结构里，

默认先想到的激活函数通常就是 ReLU。

##### 5.3 一个很直观的理解
你可以把 ReLU 理解成一个“阈值门”：
* 正值放行 ✅
* 负值直接截断为 0 ❌

这会让模型把注意力更多放在“被激活的特征”上。

例如一张特征图里，有些位置对某个卷积核响应很强，

这些位置会保留下来；

而响应弱或负向的位置，会被压掉。

#### 6. 除了 ReLU，还有哪些激活函数
**1️⃣ Sigmoid**

Sigmoid 会把输入压缩到 0 ~ 1 之间。

我们之前在 MLP 中已经学过，它更常用于：
* 二分类输出层
* 表示概率

但在卷积层中，Sigmoid 一般不是首选，因为它容易带来梯度饱和问题。

---

**2️⃣ Tanh**

Tanh 会把输入压缩到 -1 ~ 1 之间。

它比 Sigmoid 以 0 为中心一些，但同样也可能出现梯度消失。

所以在现代 CNN 的隐藏层中，Tanh 也没有 ReLU 常用。

---

**3️⃣ Leaky ReLU**

Leaky ReLU 是 ReLU 的改进版本之一。

它不是把所有负值都直接压成 0，

而是让负值保留一个很小的斜率。

例如可以理解为：
* 正数正常通过
* 负数缩小后再通过

这样做的目的是减少“神经元完全死掉”的问题。

#### 7. 卷积层、隐藏层、输出层激活函数区分⚠️

##### 7.1 卷积层中的激活函数
卷积层后面的激活函数，主要是为了：

增加非线性、帮助特征表达

所以这里通常说的是：
* ReLU
* Leaky ReLU
* 其他隐藏层激活函数

##### 7.2 全连接隐藏层中的激活函数
增加非线性、帮助特征表达

所以这里通常说的是：
* ReLU
* Leaky ReLU
* 其他隐藏层激活函数

##### 7.3 输出层中的激活函数
输出层的激活函数则是为了：

把最终特征映射成任务需要的输出形式

例如：
* 二分类：常配 Sigmoid
* 多分类：常配 Softmax
* 回归：有时不加激活函数

所以你要把这两类用途分开理解：
* 卷积层激活函数：服务于特征提取
* 输出层激活函数：服务于最终任务输出
