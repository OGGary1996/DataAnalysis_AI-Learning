## 6.4 池化层（Pooling Layer） - 平均池化（Average Pooling）

#### 1. Average Pooling 是什么

##### 1.1 基本定义
它的规则是：

>在每一个局部区域中，取所有数值的平均值作为输出。

也就是说，它不会像最大池化那样只保留最大的那个值，

而是会把这一小块区域里的信息做一次“平均概括”。

##### 1.2 一个最直观的理解
假设某个 2 × 2 的局部区域是：
```
1  3
2  6
```
那么平均池化后的输出就是：

`(1 + 3 + 2 + 6) / 4 = 3`

对每个小区域做一个局部平均。 🧾

##### 1.3 它处理的对象是什么
平均池化和最大池化一样，

处理的通常也不是原始图像，而是卷积层输出的 特征图（Feature Map）。

这些特征图中的值表示某种特征在不同位置的响应强弱。

而平均池化做的事情是：

>把局部区域中的整体响应水平概括成一个值。

#### 2. 为什么需要 Average Pooling

##### 2.1 压缩特征图尺寸
平均池化最直接的作用之一，仍然是：

缩小特征图的空间尺寸。

例如：
* 输入：28 × 28
* 经过 2 × 2 平均池化
* 输出：14 × 14

这样可以减少后续层的计算量。

##### 2.2 保留局部整体信息
和最大池化不同，平均池化不是只看最强响应，

而是更关心：

>这一整块区域总体上响应如何。

所以它更适合表达一种“整体趋势”或“平均水平”。

##### 2.3 让特征图更平滑
因为平均池化会对一个区域里的值做平均，

所以输出通常会比最大池化更平滑一些。

也就是说，它不会特别强调局部最尖锐的响应，

而是把局部信息做了一个更温和的汇总。

#### 3. Average Pooling 是如何工作的

##### 3.1 先设定池化窗口
平均池化也需要指定一个池化窗口大小。

常见的仍然是：
* 2 × 2
* 3 × 3

例如：

`2 × 2 Average Pooling`

表示每次查看一个 2 × 2 的局部区域。

##### 3.2 在局部区域中计算平均值
例如某个区域是：
```
2  4
6  8
```
那么输出就是：

`(2 + 4 + 6 + 8) / 4 = 5`

也就是说，平均池化不是选最大值，

而是把所有值都考虑进去。

##### 3.3 窗口继续滑动
池化窗口会在整张特征图上不断滑动。

每到一个位置：
* 取出当前局部区域
* 计算平均值
* 作为输出中的一个元素

最后得到一张更小的输出特征图。

#### 4. 一个完整的小例子

##### 4.1 输入特征图
假设输入特征图是：
```
1  3  2  4
5  7  1  2
6  2  8  0
4  6  3  9
```

现在使用：
* 池化窗口：2 × 2
* stride = 2

##### 4.2 第一个区域
左上角 2 × 2 区域是：
```
1  3
5  7
```
平均值是：
`(1 + 3 + 5 + 7) / 4 = 4`

##### 4.3 第二个区域
右上角 2 × 2 区域是：
```
2  4
1  2
```
平均值是：

`(2 + 4 + 1 + 2) / 4 = 2.25`

##### 4.4 第三个区域
左下角 2 × 2 区域是：
```
6  2
4  6
```
平均值是：

`(6 + 2 + 4 + 6) / 4 = 4.5`

##### 4.5 第四个区域
右下角 2 × 2 区域是：
```
8  0
3  9
```
平均值是：

`(8 + 0 + 3 + 9) / 4 = 5`

##### 4.6 最终输出
所以平均池化后的输出特征图就是：
```
4    2.25
4.5  5
```
这就是 Average Pooling 的标准计算过程。

#### 5. Average Pooling 的核心特点

##### 5.1 每个区域所有值都会参与计算
这一点和最大池化很不一样。

最大池化只保留一个最大值；

而平均池化会把局部区域中所有值都纳入考虑。

所以它保留的信息通常更加“整体化”。

##### 5.2 输出更平滑
由于是求平均，

所以平均池化后的特征图通常不会像最大池化那样突出尖锐的局部响应。

它更像是在做：

>局部平滑 + 局部概括

##### 5.3 同样没有可学习参数
平均池化通常也没有可学习参数。

它和最大池化一样，都是按照固定规则运算。

所以它本质上也是一种固定的下采样方式

##### 5.4 通常只改变空间尺寸，不改变通道数
如果输入是：

`28 × 28 × 8`

经过平均池化后，可能变成：

`14 × 14 × 8`

所以平均池化和最大池化一样，

通常压缩的是：
* 高度 H
* 宽度 W

而不是通道数 C。

#### 6. Average Pooling 和 Max Pooling 的区别

##### 6.1 最大池化：选最强
最大池化更关注：

局部区域里最强的那个响应值。

适合回答的问题是：

>“这一块里有没有明显特征？”

##### 6.2 平均池化：看整体
平均池化更关注：

这一块区域整体上的平均响应水平。

适合回答的问题是：

>“这一块整体上响应强不强？”

##### 6.3 一个直观对比例子
假设某个区域是：
```
1  1
1  8
```
那么：
* 最大池化输出：8
* 平均池化输出：(1+1+1+8)/4 = 2.75

这个例子非常典型：
* 最大池化会把最显著特征直接保留下来
* 平均池化会把它和其他值一起平均，结果更平滑

#### 7. Average Pooling 常用在什么场景
**1️⃣ 当我们希望信息更平滑时**

如果任务中希望局部信息不要太尖锐，

而是更关注整体平均趋势，

那么平均池化会比较合适。

---

**2️⃣ 在一些网络后期结构中**

虽然基础 CNN 中更常见的是 Max Pooling，

但 Average Pooling，尤其是 Global Average Pooling（全局平均池化），

在一些现代网络结构中非常常见。
