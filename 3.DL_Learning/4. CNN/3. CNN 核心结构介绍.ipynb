## 3. CNN 核心结构介绍

#### 1. CNN 是什么

##### 1.1 基本定义
CNN，全称 Convolutional Neural Network，中文叫 卷积神经网络。

它是一种特别擅长处理图像数据的神经网络结构，也可以用于视频、语音、时序信号等任务。

简单理解：
* 普通神经网络（MLP）更像是“把所有输入一股脑展开后再学习”
* CNN 更像是“先看局部，再逐步组合成整体理解” 👀

所以，CNN 非常适合图像这种有空间结构的数据。

##### 1.2 为什么图像适合用 CNN
图像不是一堆毫无关系的数字，而是一个有规律的二维网格：
* 相邻像素通常有关联
* 局部区域可以组成边缘、纹理、角点
* 更大的区域又可以组成眼睛、鼻子、轮廓、物体等

也就是说，图像天然具有：
* 局部相关性
* 空间位置关系
* 层次结构特征

而 CNN 正是专门利用这些特点设计出来的网络。

##### 1.3 一个直观比喻
我们看一张猫的图片时，不会一上来就直接“整体一次性理解”。

通常是这样的过程：
* 先看到一些边缘、线条
* 再组合成耳朵、眼睛、胡须
* 最后判断这是一只猫 🐱

CNN 的工作方式就很像这个过程：
* 前面层提取低级特征：边缘、纹理
* 中间层提取中级特征：局部形状、部件
* 后面层提取高级特征：完整目标

#### 2. CNN 主要能做什么

##### 2.1 图像分类
输入一张图片，输出它属于哪个类别。

例如：
* 猫 / 狗分类
* 手写数字识别
* 花朵分类
* 医学影像分类

这是最经典的 CNN 任务。

##### 2.2 目标检测
不仅要知道“图里有什么”，还要知道“它在哪里”。

例如：
* 检测图片中的人、车、狗
* 用框框标出目标位置

##### 2.3 图像分割
对图像中的每个像素进行分类。

例如：
* 将人和背景分开
* 医学图像中分割病灶区域

##### 2.4 其他视觉任务
* 人脸识别
* 姿态估计
* OCR 文字识别
* 视频分析
* 医学影像分析

#### 3. 为什么不能直接一直用 MLP 处理图像

##### 3.1 图像展开后会丢失空间结构
假设一张灰度图大小是 28 × 28：
* 原始图像是二维结构
* 如果送入 MLP，通常要先展开为 784 维向量

这样做的问题是：
* 原本“谁在谁旁边”的空间关系被打散了
* 模型不知道哪些像素原本是相邻的

##### 3.2 参数量会非常大
比如输入层有 784 个神经元，隐藏层有 1000 个神经元：

参数量大约就是：

`784 × 1000 + 1000 = 785000`

这还只是第一层。

如果图像更大，比如彩色图像 224 × 224 × 3，参数会爆炸式增长 💥

这会带来几个问题：
* 训练慢
* 占内存
* 更容易过拟合
* 对大图像不现实

##### 3.3 MLP 没有“局部感知”的思想
图像中的很多模式其实是局部出现的，比如：
* 一条边缘
* 一个角点
* 一块纹理

MLP 默认所有输入和输出神经元全连接，它并没有强调：
* 只看局部区域
* 同一种特征在不同位置重复出现

而 CNN 恰恰就是围绕这两个核心思想设计的。

#### 4. CNN 的核心思想

##### 4.1 局部感受野（Local Receptive Field）
CNN 不会一开始就看整张图，而是一次只看一个小区域。

比如一个 3×3 或 5×5 的小窗口：
* 先看图像左上角的一小块
* 再滑动到旁边继续看
* 最后扫描整张图

这就像我们拿着一个小放大镜在图像上移动 🔍

这样做的好处：
* 保留局部结构
* 更适合提取边缘、纹理等局部特征
* 参数更少

##### 4.2 参数共享（Weight Sharing）
在 CNN 中，同一个小滤波器会在整张图上不断滑动。

这意味着：
* 检测左上角边缘，用这个卷积核的一组参数
* 检测右下角边缘，还是用这个卷积核的这一组参数

也就是说，同一个特征检测器可以在不同位置重复使用。

好处：
* 参数数量大幅减少
* 模型更容易学习“同一种模式在不同位置都会出现”
* 更符合图像规律


##### 4.3 层次化特征提取
CNN 的每一层不直接做最终分类，而是逐层提取特征，

在下一层中使用上一层卷积层得到的结果。
* 浅层：边缘、线条、颜色变化
* 中层：纹理、角点、简单形状
* 深层：部件、目标、语义信息

这就是 CNN 非常强大的地方：

它不是手工定义特征，而是自动学习从低级到高级的特征表示。

#### 5. CNN 和MLP的关系

##### 5.1 本质上还是神经网络
CNN 虽然名字不同，但本质上仍然属于神经网络。

它也包含这些基本元素：
* 输入
* 权重参数
* 偏置
* 激活函数
* 损失函数
* 反向传播
* 优化器

CNN 只是把“层的结构设计”换成了更适合图像的方式。

##### 5.2 和 MLP 的区别主要在“如何连接”
MLP：
* 层与层之间通常是全连接
* 适合表格数据、低维特征

CNN：
* 使用卷积层代替部分全连接层
* 更适合图像这种有空间结构的数据

所以可以理解为：

CNN = 针对图像任务优化过结构的神经网络

##### 5.3 CNN + MLP

在CNN 完成了任务之后，同样还是会将图像展平送入MLP模型

这是为了综合所有的特征进行最终的分类任务

#### 6. CNN 的典型结构

##### 6.1 一个最常见的流程
一个基础 CNN 往往长这样：

`输入图像 → 卷积层 → 激活函数 → 池化层 → 卷积层（重复） → 激活函数（重复） → 池化层（重复） → 展平 → 全连接层 → 输出层`

* 卷积层：负责提取特征
* 激活函数：增加非线性表达能力
* 池化层：压缩信息、保留主要特征
* 全连接层：综合特征做最终分类
