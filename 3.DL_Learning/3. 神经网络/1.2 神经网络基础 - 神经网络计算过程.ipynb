## 1.2 神经网络初识 - 神经网络学习过程

#### 1. 神经网络是如何“学习”的？

##### 1.1 学习的本质：调整参数
在神经网络中，最核心需要学习的参数是：
* 权重（Weight，w）
* 偏置（Bias，b）

在训练开始之前，这些参数通常是随机初始化的，因此模型一开始的预测往往是比较差的。

神经网络的学习过程，本质上就是：
* 通过不断调整权重和偏置，使模型的预测结果越来越接近真实答案。

换句话说：
* 训练神经网络 = 找到一组最合适的参数（w 和 b）

##### 1.2 神经网络学习的整体流程
神经网络训练通常包含四个核心步骤：
`前向传播 → 计算损失 → 反向传播 → 参数更新`

这四个步骤会不断循环执行，直到模型训练完成。

整个过程就像：
`预测 → 发现错误 → 找到原因 → 修正参数`


#### 2. 前向传播（Forward Propagation）

##### 2.1 什么是前向传播
前向传播指的是：
数据从输入层进入神经网络，一层一层计算，最终得到预测结果的过程。

计算方向是：
`输入层 → 隐藏层 → 输出层`

例如：
```
输入一张图片：
输入：猫的图片
↓
神经网络计算
↓
输出：[0.1 , 0.9]
```

表示：
* 是狗概率：0.1
* 是猫概率：0.9

##### 2.2 前向传播内部发生了什么
在每一层神经元中，都会进行两个步骤：

**第一步：加权求和**
`z = w1x1 + w2x2 + ... + wn*xn + b`

其中：
* x：输入
* w：权重
* b：偏置

**第二步：激活函数**
`a = f(z)`

激活函数的作用是：
* 给神经网络加入非线性能力
* 最终输出 a 传递给下一层。

#### 3. 计算损失函数（Loss Function）

##### 3.1 为什么需要损失函数
神经网络需要知道：**自己预测错了多少**

损失函数就是用来衡量预测值和真实值之间的差距。

例如：
```
真实值：
y = 1
模型预测：
ŷ = 0.2
```
误差就比较大。

##### 3.2 损失函数的作用
损失函数的核心目标是：
`衡量模型预测错误的程度`

训练神经网络的目标就是：
`最小化损失函数`

#### 4. 反向传播（Backpropagation）

##### 4.1 为什么需要反向传播
前向传播只负责：计算预测结果

但神经网络还需要知道：
* 哪些参数导致了错误
* 每个参数应该如何调整

这就需要反向传播算法 - autograd自动求导+ 梯度下降

##### 4.2 反向传播的核心思想
反向传播可以理解为：
`把最终的预测误差，一层一层向前传回去。`

方向是：
`输出层 → 隐藏层 → 输入层`

在传播过程中，会计算：
`损失函数 对 每个参数 的影响程度`

这个“影响程度”就是：
`梯度（Gradient）`

##### 4.3 梯度的直观理解
梯度可以理解为：
`如果改变这个参数一点点，损失会如何变化`

例如：

如果：
* 参数增加一点
* 损失变大

说明这个方向是错的。

如果：
* 参数减少一点
* 损失变小

说明应该往这个方向更新。

##### 4.4 反向传播的作用总结
反向传播的本质就是：
`计算损失函数对每个参数的梯度`

有了梯度，我们就知道：
* 参数应该往哪个方向调整
* 调整多少

#### 5. 参数更新（梯度下降）

##### 5.1 梯度下降的基本思想
参数更新最经典的方法是：
`Gradient Descent（梯度下降）`

更新公式：
`新参数 = 旧参数 - 学习率 × 梯度`

其中：
* 梯度：告诉我们方向
* 学习率：决定更新幅度

##### 5.2 学习率（Learning Rate）
学习率决定：
`每次更新参数时迈多大的步子`

如果学习率：
* 太大：
    * 参数更新过猛
    * 可能无法收敛
* 太小：
    * 学习速度很慢
    * 训练时间很长

所以学习率是一个非常重要的超参数。

##### 5.3 训练的循环过程
神经网络训练通常会不断重复以下过程：
```
输入数据
↓
前向传播
↓
计算损失
↓
反向传播
↓
更新参数
↓
重复训练
```
这个循环通常会持续：
* 几十次、几百次甚至几千次

直到模型收敛。


#### 6. 神经网络中的重要概念

##### 6.1 参数（Parameters）
参数是：
`模型在训练过程中学习得到的数值`

主要包括：
* 权重（Weights）
* 偏置（Bias）

##### 6.2 超参数（Hyperparameters）
超参数是：
`在训练之前由我们人为设定的参数`
* learning rate
* batch size
* epoch
* 神经元数量
* 网络层数


##### 6.3 Epoch、Batch、Iteration
这三个概念非常常见。
* Epoch
    * 整个训练集被完整训练一次。
* Batch
    * 一次输入模型的一小部分数据。
* Iteration
    * 一次参数更新

`1 Epoch = 多个 Batch = 多次 Iteration`

例如：

训练集有 1000 个样本：
`batch size = 100`

那么：
`1 epoch = 10 batches = 10 iteration`

#### 7. 过拟合与泛化能力
##### 7.1 什么是过拟合（Overfitting）
模型在训练数据上表现很好，但在新数据上表现很差。

表现为：
* 训练集表现很好
* 测试集表现很差

原因通常是：
`模型把训练数据“记住了”，而不是学到了规律。`

##### 7.2 什么是泛化能力（Generalization）
模型在未见过的新数据上的表现能力

好的模型应该：
* 在训练集上表现好
* 在测试集上也表现好

#### 8. 神经网络为什么被称为“黑箱模型”？

##### 8.1 什么是黑箱模型
`输入和输出是清楚的，但内部计算过程难以解释。`

例如：

输入：
`MRI影像`

输出：
`患病概率 0.87`

但中间到底为什么得到这个结果，很难完全解释。

##### 8.2 为什么神经网络容易成为黑箱
* 网络层数很多
* 参数数量巨大
* 内部特征自动学习

因此很难用人类语言解释每一步。

##### 8.3 黑箱并不代表不好
虽然解释性较弱，但神经网络在很多任务上表现非常优秀，例如：
* 图像识别
* 语音识别
* 医学影像诊断
* 自动驾驶

所以实际应用中往往在：`效果  vs  可解释性`之间进行权衡。


#### 9. 神经网络与之前学习模型的关系

##### 9.1 神经网络与线性回归
一个神经元的计算公式：
`z = wx + b`

这与线性回归模型非常相似。

因此可以理解为：**神经网络是线性模型的扩展。**

##### 9.2 神经网络与逻辑回归
逻辑回归的形式是：
```
z = wx + b
ŷ = sigmoid(z)
```
这其实就是：
`一个没有隐藏层的神经网络`

结构：
`输入层 → 输出层`
