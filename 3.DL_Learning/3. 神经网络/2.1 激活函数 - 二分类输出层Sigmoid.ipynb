## 2.1 激活函数 - Sigmoid

#### 1. 什么是 Sigmoid 激活函数

Sigmoid 激活函数的公式为：

```
sigma(x) = 1/ (1 + e^(-x))
```
其中：
* e 是自然常数（约等于 2.718）
* x 是神经元的输入值

```
💡常用于二分类任务 (Binary Classification)
* 结构：输出层只有 1个 神经元。
* 激活函数：使用 Sigmoid 函数。它将输出值压缩到 [0,1] 之间，表示“属于正类的概率”。
* 判定逻辑：通常以 0.5 为阈值。输出 >0.5 为正类，<0.5 为负类。
* 损失函数： 二元交叉熵损失 (Binary Cross Entropy, BCE)。
```

##### 1.1 Sigmoid 函数形状
Sigmoid 函数的图像呈现：S 型曲线

特点：
````
输入	输出
x → -∞	输出 → 0
x = 0	输出 = 0.5
x → +∞	输出 → 1
````


##### 1.2 Sigmoid 的输出范围
Sigmoid 的输出范围是：
`(0 , 1)`

注意：
**不是包含 0 和 1，而是：**
`0 < σ(x) < 1`


##### 1.3 为什么 Sigmoid 常用于概率
因为它的输出范围在：`0 ~ 1`;这与概率的范围完全一致。

因此 Sigmoid 常用于：二分类问题

例如：
`P(y = 1 | x)`

输出解释为：
```
输出	含义
0.9	非常可能是正类
0.5	不确定
0.1	非常可能是负类
```

#### 2. Sigmoid 的直观理解
理解 Sigmoid 可以用一个简单的逻辑：
**把任意实数映射到 0~1 之间。**

例如：
```
输入 x	Sigmoid(x)
-10	≈ 0
-2	0.12
0	0.5
2	0.88
10	≈ 1
```
所以 Sigmoid 就像：一个“概率压缩函数”,把任意数压缩到概率范围。

#### 3. Sigmoid 的数学性质（重要）
Sigmoid 有一个非常重要的性质：

##### 3.1 Sigmoid 的导数
Sigmoid 的导数非常漂亮：
`sigma'(x) = sigma(x) * (1 - sigma(x))`

也就是说：
`导数 = 输出 × (1 - 输出)`

如果：
`σ(x) = 0.8`

那么：
`σ'(x) = 0.8 × 0.2 = 0.16`


##### 3.2 为什么这个性质重要
因为在神经网络训练中，需要计算梯度：

Sigmoid 的导数计算非常简单，因此在早期神经网络中非常流行。

#### 4. Sigmoid 在神经网络中的作用

##### 4.1 二分类输出层
例如：
`输入 → 神经网络 → Sigmoid → 概率`

输出：
`P(y = 1)`

再通过阈值判断：
* >0.5 → 正类
* <0.5 → 负类

##### 4.2 逻辑回归
逻辑回归的核心公式：
```
z = wx + b
y = sigmoid(z)
```
所以：
**⭐ 逻辑回归本质上就是一个使用 Sigmoid 的单层神经网络。**

#### 5. Sigmoid 的缺点（非常重要）

##### 5.1 梯度消失（Vanishing Gradient）
Sigmoid 在两端区域会变得非常平缓：
```
x 很大 → σ(x) ≈ 1
x 很小 → σ(x) ≈ 0
```

这时导数：
`σ'(x) ≈ 0`

这意味着：梯度非常小。

如果网络层数很多：
`小梯度 × 小梯度 × 小梯度`
`最终可能变成：0`

结果就是：

前面的层几乎无法更新参数

这就是：梯度消失问题。

##### 5.2  输出不是零中心
Sigmoid 的输出范围是：0-1，所有值都是正数。

这会导致：
* 梯度更新方向不稳定
* 训练速度变慢

相比之下：tanh，输出范围 -1 - 1，更适合训练。

##### 5.3 计算成本较高
Sigmoid 包含：
`e^x`

指数计算比：ReLU更小号计算资源

虽然现在硬件很强，但在大型网络中仍然有影响。

#### 6. Sigmoid 与其他激活函数的关系
Sigmoid 在深度学习发展中具有重要历史地位。

发展大致经历：
`Sigmoid → Tanh → ReLU → ReLU 变体`

当前最常用的是：
* ReLU
* Leaky ReLU
* GELU
* Swish

但 Sigmoid 仍然非常重要，因为：
* 二分类输出层仍然使用
* Logistic Regression 使用
* 概率建模使用
