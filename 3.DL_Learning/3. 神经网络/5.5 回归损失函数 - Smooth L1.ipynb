## 5.5 回归损失函数 - Smooth L1 
在回归任务中，我们已经学习了：
* MSE Loss（L2 Loss）
* MAE Loss（L1 Loss）

但是这两种损失函数各自都有缺点：
```
损失函数	优点	缺点
MSE L2	梯度平滑	对异常值非常敏感
MAE L1	对异常值稳定	梯度不稳
```

因此，在很多实际任务中，人们使用一种折中方法：

`Smooth L1 Loss`

它结合了：
* L1 Loss
* L2 Loss


#### 1. 什么是 Smooth L1？
Smooth L1 Loss 是一种 **分段损失函数（Piecewise Loss Function）。**

核心思想是：
```
误差小 → 使用 MSE L2 Loss（平方误差）
误差大 → 使用 MAE L1 Loss（绝对误差）
```

也就是说：
* 小误差 → 平滑优化
* 大误差 → 防止异常值影响

##### 1.1 为什么需要这种设计？
我们来看两种情况：

**情况1：误差很小** 

`|y - ŷ| 很小`

此时如果使用L1 Loss：
* 梯度不稳定

使用：L2 Loss

优点：
* 梯度平滑
* 收敛速度快

**情况2：误差很大**

`|y - ŷ| 很大`

此时如果使用 L2 Loss：
* 异常值会被平方放大

使用：L1 Loss

#### 2. Smooth L1 的数学公式
Smooth L1 的公式为：
```
Loss(x < 1) = 0.5 * x^2
Loss(x ≥ 1) = |x| - 0.5
```

`x = y - ŷ`

Smooth L1 在 小误差区域像 MSE，在大误差区域像 MAE。

##### 2.1 小误差区域
如果：
`|x| < 1`

使用：

` x^2 * 0.5 `

这相当于：L2 Loss

##### 2.2 大误差区域
如果：
`|x| ≥ 1`

使用：

`|x| - 0.5`

这相当于：L1 Loss

#### 3. Smooth L1 的直观理解
Smooth L1 可以理解为：
```
小误差 → 平滑惩罚
大误差 → 线性惩罚
```
因此：
**它既有 L2 的平滑性，又有 L1 的鲁棒性。**

#### 4. Smooth L1 与 MSE / MAE 的区别
现在可以把三种回归损失函数放在一起理解：
```
Loss	特点
MSE	平滑，但对异常值敏感
MAE	鲁棒，但梯度不平滑
Smooth L1	两者折中
```
1️⃣ 对异常值的敏感程度

`MSE  >  Smooth L1  >  MAE`

2️⃣ 梯度稳定性

`MSE > Smooth L1 > MAE`

#### 5. PyTorch 实现 Smooth L1
在 PyTorch 中：

`nn.SmoothL1Loss()`

##### 5.1 定义损失函数
`criterion = nn.SmoothL1Loss()`

##### 5.2 输入格式
与 MSE 相同：

`loss = criterion(pred, target)`

形状：
```
pred   = (batch_size , output_size)
target = (batch_size , output_size)
```

##### 5.3 示例代码

````python
import torch
import torch.nn as nn

criterion = nn.SmoothL1Loss()

pred = torch.tensor([[2.5], [3.2], [4.1]])

target = torch.tensor([[3.0], [3.0], [4.0]])

loss = criterion(pred, target)

print(loss)
````

#### 6. Smooth L1 的应用场景
Smooth L1 常用于：

1️⃣ 目标检测（Object Detection）

例如：
* Fast R-CNN
* Faster R-CNN
* Mask R-CNN

在这些模型中：

`bounding box regression`

通常使用：

`Smooth L1 Loss`
原因是：
* 目标检测中的 bounding box 偏差可能非常大。
* 使用 MSE 会导致：异常值经过放大 非常敏感

2️⃣ 强化学习
在 `Deep Q Network (DQN)` 中也经常使用：

`Huber Loss`

实际上：

`Huber Loss ≈ Smooth L1`
