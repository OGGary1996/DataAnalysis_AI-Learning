## 5.4 回归损失函数 - 平均绝对误差（MAE Loss / L1 Loss）
在回归任务中，我们已经学习了：
* MSE（Mean Squared Error）

但是 MSE 有一个明显的问题：

`对异常值（outliers）非常敏感`

因此在一些任务中，人们会使用另一种损失函数：

`MAE（Mean Absolute Error）`

也叫：

`L1 Loss`

#### 1. 什么是 MAE？
MAE 的核心思想非常简单：
`计算预测值与真实值之间的绝对误差的平均值`

例如：

真实值：

`y = [3, 5, 2]`

预测值：

`ŷ = [2.5, 5.3, 1.8]`

误差：
```
|3 - 2.5| = 0.5
|5 - 5.3| = 0.3
|2 - 1.8| = 0.2
```
平均：

`MAE = (0.5 + 0.3 + 0.2) / 3`

得到：

`MAE = 0.333`

#### 2. MAE公式
MAE 的数学公式为：

`Loss = SUM( |y_i - ŷ_i| ) / n`

其中：
* y_i → 真实值
* ŷ_i → 预测值
* n → 样本数量

##### 2.1 为什么使用绝对值？
如果直接计算：

`y - ŷ`

可能出现：
```
+10
-10
```

平均后可能变成：

`0`

但实际误差并没有消失。

因此需要：绝对值来表示误差的真实大小。

#### 3. MAE 的核心特点
MAE 的一个重要特点是：

对异常值不敏感

我们来看一个例子。

##### 3.1 示例
真实值：

`[10, 12, 11, 100]`

预测值：
`[9, 11, 10, 12]`

最后一个值是 异常值。

##### 3.2 使用 MSE
误差平方：

`(100 - 12)^2 = 7744`

这会导致：

`loss 非常大`

模型会过度关注这个异常值。

#### 3.2 使用 MAE
误差：

`|100 - 12| = 88`

虽然也大，但不会被平方放大。

#### 4. MAE 与 MSE 的区别
```
Loss	公式	   特点
MSE    (y-ŷ)²	对异常值敏感
MAE	   |y-ŷ|    对异常值鲁棒
```

##### 4.1 梯度区别
MSE 的梯度：

`2(y - ŷ)`

MAE 的梯度：

`sign(y - ŷ)`

也就是说：
* MSE → 梯度随误差变大
* MAE → 梯度大小基本固定

#### 5. MAE 的直观理解
MAE 可以理解为：

`预测值与真实值之间的平均距离`

如果：
* 预测值接近真实值

那么：
* 预测值偏差很大

那么：
* loss 很大

#### 6. PyTorch 实现 MAE
在 PyTorch 中：

`nn.L1Loss()`

就是 MAE Loss。

##### 6.1 定义损失函数
`criterion = nn.L1Loss()`

##### 6.2 输入格式
与 MSE 完全相同：

`loss = criterion(pred, target)`

形状：
```
pred   = (batch_size , output_size)
target = (batch_size , output_size)
```

##### 6.3 示例代码

````python
import torch
import torch.nn as nn

criterion = nn.L1Loss()

pred = torch.tensor([[2.5], [3.2], [4.1]])
target = torch.tensor([[3.0], [3.0], [4.0]])

loss = criterion(pred, target)

print(loss)
````

#### 7. MAE 的优缺点
✅ 优点
* 对异常值更鲁棒
* 不会过度惩罚大误差

🙅 缺点
MAE 的梯度在：

`误差 = 0`

处 不可导（不光滑）。

因此：
* 梯度优化不如 MSE 平滑
* 收敛速度可能较慢
