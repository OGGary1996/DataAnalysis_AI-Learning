## 5.3 损失函数 - 均方误差（MSE Loss）
在机器学习和深度学习中，如果任务是 预测连续数值，通常称为：

`回归任务（Regression）`

例如：
* 房价预测
* 温度预测
* 股票价格预测
* 销售额预测

在回归任务中最常使用的损失函数是：

`MSE（Mean Squared Error）`

#### 1. 什么是回归问题？
回归问题指的是：

`模型需要预测一个连续数值。`

例如：

预测房价：
```
特征	房价
面积	350000
面积	420000
面积	280000
```
模型的任务是：

`输入特征 → 预测一个连续数值`

例如：

`pred = 365000`

##### 1.1 模型输出是什么？
在回归问题中，模型的输出通常是：一个实数

例如：

`pred = 3.5`

或：

`pred = 250000`

与分类问题不同：

⚠️ 回归输出不表示概率

因此：
* 不需要 Softmax
* 不需要 Sigmoid

通常最后一层就是：

`Linear`

✨不需要额外的激活函数

##### 1.2 为什么需要损失函数？
模型预测的值通常不会完全等于真实值。

例如：

真实房价：

`y = 400000`

模型预测：

`ŷ = 370000`

我们需要一个函数来衡量：

`预测值 vs 真实值`

之间的差距。

这就是：

`损失函数（Loss Function）`

#### 2. MSE公式
MSE 的公式是：

`MSE = (SUM(y_i - ŷ_i)^2) / n`

其中：
* y_i → 真实值
* ŷ_i → 预测值
* n → 样本数量

##### 2.1 为什么要平方？
平方有两个重要作用：

**① 消除正负误差**

如果直接计算：

`y - ŷ`

可能出现：
```
+10
-10
```
平均后可能变成：
`0`

但实际误差并没有消失。

因此需要：平方处理

**② 放大大的错误**

平方会放大较大的误差：
```
误差	平方
1	1
2	4
5	25
```

因此：

大的预测错误会受到更大的惩罚。

#### 3. MSE 的核心思想
MSE 的核心思想可以理解为：

计算预测值与真实值之间的平均平方距离。

如果：
* 预测值接近真实值

那么：
* loss 很小

如果：
* 预测值与真实值差距很大

那么：
* loss 很大

##### 3.1 一个简单例子
假设有 3 个样本：

真实值：

`y = [3, 5, 2]`

模型预测：

`ŷ = [2.5, 5.3, 1.8]`

计算误差：
```
(3 - 2.5)^2 = 0.25
(5 - 5.3)^2 = 0.09
(2 - 1.8)^2 = 0.04
```

求平均：

`MSE = (0.25 + 0.09 + 0.04) / 3`

得到：

`MSE = 0.1267`

#### 4. MSE 在神经网络中的计算流程在神经网络训练中：
```
输入特征
↓
神经网络
↓
预测值 ŷ
↓
计算 MSE
↓
反向传播更新参数
```
也就是说：

`Loss = MSE(pred , target)`

##### 4.1 输出层结构
在回归问题中，输出层通常是：

`Linear`

例如：

`Linear → MSELoss`

##### 4.2 注意：
与分类任务不同：
* 不需要 Softmax
* 不需要 Sigmoid

因为：输出不是概率

##### 5. PyTorch 实现 MSE

##### 5.1 定义损失函数
`criterion = nn.MSELoss()`

##### 5.2 输入格式
MSELoss 需要两个输入：

`loss = criterion(pred, target)`

pred（模型预测）

形状：

`(batch_size , output_size)`

例如：

`(32,1)`

表示：
* 32 个样本
* 每个样本预测 1 个数值

target（真实值）

形状：

`(batch_size , output_size)`

例如：

`(32,1)`

注意：

⚠️ 与 CrossEntropy 不同：

target 不是类别编号

而是：

真实数值

##### 5.3 示例代码

````python
import torch
import torch.nn as nn

# 定义损失函数
criterion = nn.MSELoss()

# 模型预测值
pred = torch.tensor([[2.5], [3.2], [4.1]])

# 真实值
target = torch.tensor([[3.0], [3.0], [4.0]])

loss = criterion(pred, target)

print(loss)
````

##### 6. 为什么回归任务通常使用 MSE？
原因主要有三个：

1️⃣ 数学性质好

MSE 是：

`连续可导函数`

这对于：梯度下降非常友好。

2️⃣ 对大误差更敏感

平方会放大大误差：

`大错误 → 更大惩罚`

这可以促使模型快速修正严重错误。

3️⃣ 计算简单

MSE 的计算非常简单：

`(真实值 - 预测值)^2`

因此计算效率高。
