## 2.3 激活函数 - Tanh

#### 1. 什么是 Tanh 激活函数
Tanh 是 Sigmoid 的改进版本之一，全称是：`Hyperbolic Tangent（双曲正切函数）`

数学公式：
`tanh(x) = (e^x - e^(-x)) / (e^x + e^(-x)) `

在神经网络中，Tanh 的作用和 Sigmoid 一样：为神经网络引入非线性能力

##### 1.1 Tanh 的函数形状
Tanh 的函数图像同样是 S 型曲线：
```
输入	输出
x → -∞	tanh(x) → -1
x = 0	tanh(x) = 0
x → +∞	tanh(x) → 1
```


##### 1.2 Tanh 的输出范围
Tanh 的输出范围是：
`(-1 , 1)`

也就是说：
`-1 < tanh(x) < 1`

这一点和 Sigmoid 不同：
```
激活函数	输出范围
Sigmoid	(0 , 1)
Tanh	(-1 , 1)
```

#### 2. Tanh 的导数
Tanh 的导数公式非常漂亮：
`tanh'(x) = 1 - tanh(x)^2`

如果：
`tanh(x) = 0.6`

那么导数：
`1 - 0.36 = 0.64`

相比 Sigmoid：
`σ'(x) = σ(x)(1-σ(x))`

Tanh 的梯度通常更大一些，因此：反向传播更容易进行。


#### 3. Tanh 与 Sigmoid 的关系
一个非常重要的知识点是：
`Tanh 可以看作是 Sigmoid 的变形。`

数学关系是：
`tanh(x) = 2*sigmoid(2x) - 1`

也就是说：
`Sigmoid → 向上平移 → 拉伸 → Tanh`

因此：
* 两者形状类似
* 但 Tanh 以 0 为中心


#### 4. 为什么 Tanh 比 Sigmoid 更好
在深度学习的发展过程中，人们发现：
`Tanh 通常比 Sigmoid 更适合作为隐藏层激活函数。`

##### 4.1 输出是零中心（Zero-centered）
Sigmoid 的输出范围：
`0 ~ 1`
所有输出都是 正数。

这会导致一个问题：
**梯度更新方向不稳定。**

而 Tanh 的输出范围：
`-1 ~ 1`

输出是 以 0 为中心。
* 梯度更新更加稳定
* 训练速度更快

##### 4.2 梯度更大（训练更快）
在 Sigmoid 中，当：
`|x| 很大`

Sigmoid 的导数会变得非常小：
`≈ 0`

这会导致：
**梯度消失问题。**

Tanh 也存在类似问题，但通常：
* 梯度范围更大
* 训练效果更好

#### 5. Tanh 的优点
```
优点	解释
Zero-centered	输出范围 (-1,1)
梯度更大	训练更稳定
表达能力强	非线性更明显
```

因此在早期深度学习中：
Tanh 经常作为隐藏层激活函数。


#### 6. Tanh 的缺点

##### 6.1 仍然存在梯度消失
当：
`|x| 很大`

Tanh 会进入 饱和区：
```
tanh(x) ≈ 1
tanh(x) ≈ -1
```

此时导数：
`≈ 0`

这意味着：
* 梯度会变得很小
* 深层网络训练困难

##### 6.2 计算仍然比较复杂
Tanh 的计算包含：
`e^x`
指数运算。

相比之下：ReLU，只需要简单的：max(0,x)，因此计算成本更低。

#### 7. Tanh 在神经网络中的应用
在现代深度学习中：
```
位置	是否常用
隐藏层	较少
RNN 网络	常见
输出层	很少
```

例如：
在 RNN（循环神经网络） 中：
隐藏状态更新
常常使用 Tanh。


#### 8. Sigmoid 与 Tanh 对比总结
```
特性	Sigmoid	Tanh
输出范围	(0,1)	(-1,1)
是否零中心	否	是
梯度大小	较小	较大
梯度消失	严重	较轻
隐藏层使用	不推荐	可以
```
总结一句话：

**⭐ Tanh 是 Sigmoid 的改进版本。**
