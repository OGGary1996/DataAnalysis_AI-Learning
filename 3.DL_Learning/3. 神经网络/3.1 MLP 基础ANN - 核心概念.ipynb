## 3.1 MLP基础ANN - 核心概念

#### 1. 什么是 MLP（多层感知机）
MLP 是：`最基础、最经典的神经网络结构。`

MLP 是 ANN 中最基本的一种结构，也常被称为前馈神经网络 (Feedforward Neural Network)。

它的本质就是：
`⭐ 由多层神经元组成的前馈神经网络`

结构和神经网络相同，通常为：`输入层 → 隐藏层 → 隐藏层 → ... → 输出层`

例如：
```
x1   x2   x3
 │    │    │
 └────┴────┘
      ↓
   Hidden Layer
      ↓
   Hidden Layer
      ↓
   Output Layer
```
MLP 的核心思想：
`通过多层线性变换 + 激活函数，学习复杂的非线性关系。`

#### 2. MLP 的基本结构

##### 2.1 输入层（Input Layer）
输入层的作用：**接收原始数据**

假设数据有 3 个特征：`x = [x1, x2, x3]`

输入层就有：`3 个神经元`

例如：
```
特征	示例
年龄	35
收入	60000
信用评分	720
```
输入层不会进行复杂计算，它只是：
**⭐ 把数据传给下一层**

##### 2.2 隐藏层（Hidden Layer）
**神经网络学习数据规律的核心部分**

每个神经元都会进行：`加权求和 + 激活函数`

数学形式：
```
z = Wx + b
a = f(z)
```
例如：
```
Input (3)
      ↓
Hidden Layer (5 neurons)
```
表示：`3个输入 → 5个神经元`

每个神经元都会：
`w1x1 + w2x2 + w3x3 + b`

然后通过激活函数：
`ReLU / Tanh / Sigmoid`

##### 2.3 输出层（Output Layer）
**给出最终预测结果**

不同任务的输出层不同：
```
任务	输出层
回归	1个神经元
二分类	1个神经元 + Sigmoid
多分类	n个神经元 + Softmax
```

例如：手写数字识别：

`10个类别`

输出层：

`10个神经元`

输出：

`[0.01,0.02,0.90,...]`

表示每个数字的概率。


#### 3. 为什么叫“多层感知机”
MLP 的名字来自两个词：

1️⃣ Perceptron（感知机）

感知机是：**最早的神经元模型**

公式：

`y = sign(wx + b)`

这是最早的分类器之一。

2️⃣ Multi-Layer（多层）

早期感知机只有：

`输入层 → 输出层`

但这种模型只能解决：⭐ 线性可分问题

后来人们发现，如果增加隐藏层：

`Input → Hidden → Output`

就能解决复杂问题。

因此：

Multi-Layer Perceptron

就是：多个感知机层叠组成的网络

#### 4. MLP 的信息流动（前向传播）
MLP 是一种：

**⭐ 前馈神经网络（Feedforward Neural Network）**

意思是：数据只向前流动

结构：

`Input → Hidden1 → Hidden2 → Output`

不会出现循环

#### 5. MLP 的数学表达
输入：

`x ∈ R^d`

第一层：
```
z1 = W1x + b1
a1 = f(z1)
```

第二层：
```
z2 = W2a1 + b2
a2 = f(z2)
```

输出层：

`y = W3a2 + b3`

可以写成：

`y = f(W3 f(W2 f(W1x + b1) + b2) + b3)`

这就是：**⭐ 神经网络的层级函数组合**

#### 6. MLP 的核心能力
MLP 最重要的能力是：**逼近任意函数**

这来自一个重要理论：

`⭐ Universal Approximation Theorem`

只要隐藏层足够大，MLP 可以逼近任何连续函数。

MLP 可以学习：
* 图像规律
* 语音模式
* 复杂数据关系


#### 7. MLP 的典型应用
1️⃣ 回归问题

例如，房价预测

输入：

`面积、位置、房龄`

输出：

`房价`

2️⃣ 分类问题

例如：垃圾邮件分类

输入：

`邮件文本特征`

输出：

`垃圾 / 非垃圾`

3️⃣ 特征学习

在很多深度学习模型中：

MLP 用于：

特征变换

例如：Transformer 中就存在 MLP block


#### 8. MLP 在深度学习中的地位
```
线性回归
      ↓
逻辑回归
      ↓
单层感知机
      ↓
MLP
      ↓
CNN / RNN / Transformer
```

⭐ MLP 是所有深度学习模型的基础结构。
