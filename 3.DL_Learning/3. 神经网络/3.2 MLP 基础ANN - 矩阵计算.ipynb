## 3.2 MLP 基础ANN - 矩阵计算

在前一节中，我们已经理解了 MLP 的结构：

`Input → Hidden → Output`

并且知道每个神经元在做的事情是：
```
z = w1x1 + w2x2 + ... + wn*xn + b
a = activation(z)
```

但如果神经元很多，例如：
* 输入 100 个特征
* 隐藏层 256 个神经元
* 批量输入 128 条数据

如果逐个神经元计算会非常复杂。

因此在深度学习中：**⭐ 所有计算都会转换为矩阵运算。**

#### 1. 单个神经元和单样本的计算
**💡 神经网络本质是大量线性计算**

一个神经元，一条输入样本：

`z = w1x1 + w2x2 + w3x3 + ... + wn*xn + b`

如果写成向量形式：
`z = w·x + b`

也就是：
`z = w^T x + b`

这其实就是：**⭐ 向量点积（dot product）**

#### 2. 多神经元和单样本计算（一层神经网络）

##### 💡 多个神经元可以一起计算

假设现在：
* 输入是 1 条样本
* 这条样本有 n 个特征
* 当前这一层有 4 个神经元

那么 4 个神经元会分别对同一条样本进行计算：
```
z_1 = w11x1 + w12x2 + w13x3 + ... + w1n*xn + 1b
z_2 = w21x1 + w22x2 + w23x3 + ... + w2n*xn + 2b
z_3 = w31x1 + w32x2 + w33x3 + ... + w3n*xn + 3b
z_4 = w41x1 + w42x2 + w43x3 + ... + w4n*xn + 4b
```

如果把 4 个神经元一起计算，就可以写成矩阵形式。

##### 2.1 输入向量（Input Vector）
假设一个样本有 n 个特征,在 PyTorch 中，通常把 1 条样本写成 1 行：

`x = [x1 x2 x3 ... xn]`

形状：

`(1 × n)`

表示：
* 一个样本
* n个特征

##### 2.2 权重矩阵（Weight Matrix）
隐藏层有 4 个神经元，每个神经元都有 n 个权重：
```
Neuron1 → w11 w12 w13 ... w1n
Neuron2 → w21 w22 w23 ... w2n
Neuron3 → w31 w32 w33 ... w3n
Neuron4 → w41 w42 w43 ... w4n
```

可以写成⭐ 权重矩阵：
```
W =
[w11 w12 w13 ... w1n
 w21 w22 w23 ... w2n
 w31 w32 w33 ... w3n
 w41 w42 w43 ... w4n]
```

矩阵形状：
`(4 × n)`

表示：
* 4 个神经元
* 每个神经元接收 n 个输入特征

##### 2.3 偏置向量（Bias Vector）
每个神经元都有一个 bias：
```
b =
[b1
 b2
 b3
 b4]
```

形状：
`(4 × 1)`

也可以把它理解成有 4 个元素的一维向量，PyTorch 会自动广播。

##### 2.4 线性计算
在 PyTorch 常见写法中，一层神经网络的线性计算写成：
`Z = XW^T + b`

⚠️注意：
* 权重矩阵W.shape = (4, n)
* 特征矩阵X.shape = (4,n)
* 无法直接相乘，所以需要将W^T转置处理
* 在pytorch框架内部，会自动将W进行转置处理
* 并且矩阵乘法的顺序是 X · W.T

转置后矩阵乘法：

`(1 x n) · (n x 4) = (1 x 4)`

结果：

`Z = [z1 z2 z3 z4]`

形状：
`（1 x 4）`

表示：
* 1 条样本
* 经过这一层后，得到 4 个神经元的输出

##### 2.5 激活函数
再通过激活函数：

`A = activation(Z)`

例如：

`A = ReLU(Z)`

结果：

`A = [a1 a2 a3 a4]`

形状仍然是：

`(1 × 4)`

这就是这一层隐藏层的输出。

#### 3. 多层神经网络的矩阵计算

现在我们把网络扩展成多层。

假设网络结构如下：

`Input (n) → Hidden (4神经元) → Output (2神经元)`

##### 3.1 隐藏层计算
输入：

`X.shape = (1, n)`

隐藏层权重：

`W1.shape = (4, n)`

隐藏层偏置：
`b1.shape = (1, 4)`

那么隐藏层计算为：
```
Z1 = XW1^T + b1
A1 = activation(Z1)
```

维度变化：

`(1 × n) · (n × 4) + (1 × 4) = (1 × 4)`

所以：

`A1.shape = (1, 4)`

##### 3.2  输出层计算
现在输出层接收隐藏层输出 A1 作为输入。

输入和隐藏层的输出A1保持一致

`A1.shape = (1, 4)`

隐藏层计算，2个神经元：
`W2.shape = (2, 4)`

第二层偏置：

`b2.shape = (1, 2)`

那么第二层计算为：
```
Z2 = A1W2^T + b2
A2 = activation(Z2)
```

维度变化：

`(1 × 4) · (4 × 2) + (1 × 2) = (1 × 2)`
所以：

`A2.shape = (1, 2)`

表示最终输出层有 2 个输出值。


##### 3.3 多层网络的整体形式
所以，一个两层 MLP 可以写成：
```
A1 = activation(XW1^T + b1)
A2 = activation(A1W2^T + b2)
```

合起来写就是：

`A2 = activation(activation(XW1^T + b1)W2^T + b2)`

这就是多层神经网络前向传播的矩阵形式。

#### 4. 批量数据的矩阵计算（Batch Computation）
真实训练时，我们通常不会只输入 1 条样本，而是一次输入一个 batch。
假设：
* batch size = 32
* 每条样本有 n 个特征
* 当前层有 4 个神经元

##### 4.1 输入矩阵
这时输入矩阵 X 写成：
```
X =
[x11 x12 x13 ... x1n
 x21 x22 x23 ... x2n
 ...
 x32,1 x32,2 x32,3 ... x32,n]
```

形状：

`(32 × n)`

表示：
* 32 条样本
* n 个特征

##### 4.2 权重矩阵
这一层有 4 个神经元，所以：

`W.shape = (4, n)`

偏置：

`b.shape = (1, 4)`

##### 4.3 线性计算
线性层计算：

`Z = XW^T + b`

维度变化：

`(32 × n) · (n × 4) + (1 × 4) = (32 × 4)`

所以：

`Z.shape = (32, 4)`

表示：
* 32 条样本
* 每条样本都得到 4 个神经元输出

##### 4.4 激活函数
再通过激活函数：

`A = activation(Z)`

结果：
`A.shape = (32, 4)`

这就是这一层对一个 batch 的输出结果。

#### 5. MLP 的完整前向传播流程（Forward Pass）

现在把整个流程串起来。

假设网络结构：

`Input (n) → Hidden (4) → Output (k)`

并且一次输入一个 batch：
`X.shape = (batch_size, n)`

那么完整前向传播可以写成：
```
X
↓
Z1 = XW1^T + b1
↓
A1 = ReLU(Z1)
↓
Z2 = A1W2^T + b2
↓
A2 = Softmax(Z2)
↓
Output
```

##### 5.1 隐藏层
```
X.shape = (batch_size, n)
W1.shape = (4, n)
b1.shape = (1, 4)

Z1 = XW1^T + b1
A1 = ReLU(Z1)
```

结果：

`A1.shape = (batch_size, 4)`

##### 5.2 输出层
```
W2.shape = (k, 4)
b2.shape = (1, k)

Z2 = A1W2^T + b2
A2 = Softmax(Z2)
```

结果：

`A2.shape = (batch_size, k)`

##### 5.3 最核心的一句话
在 PyTorch 的写法里，神经网络每一层的本质都是：
```
Z = XW^T + b
A = activation(Z)
```

其中：
* X：输入矩阵，样本按行排列
* W：权重矩阵，形状是 (out_features, in_features)
* W^T：转置后才能与 X 相乘
* b：偏置，通常按输出维度展开并自动广播
* A：激活函数输出

#### 6. PyTorch 中 Linear 层也遵循这个规律
在 PyTorch 中：

`nn.Linear(in_features, out_features)`

例如：

`nn.Linear(3,4)`

表示：
* 3个输入特征（神经元）
* 4个输出特征（神经元）
* 没有隐藏层

##### ⚠️ 非常重要：

nn.Linear(3,4) 内部参数 shape：
```
weight: (4,3)
bias:   (4,)
```

注意：
* weight权重矩阵中的特征个数应该放在和X特征相同的维度，个数匹配
* 比如三个特征的X（batch_size，3）， 对应3个权重的weight（n，3）
* 又因为没有隐藏层，所以神经元个数就是输出层神经元个数


##### 🧩 为什么这么设计？

如果：

`nn.Linear(3, 2)`

意味着：
* 输入 3 个特征
* 输出 2 个神经元

weight shape 会是：

`(2, 3)`

解释：
* 第 1 行 → 第 1 个神经元的权重，
    * 与X的特征数量保持一致为3
* 第 2 行 → 第 2 个神经元的权重
    * 与X的特征数量保持一致为3

#### 7. 七、本节核心总结
MLP 的所有计算都可以写成：
```
Z = XW + b
A = activation(Z)
```

⭐ 神经网络本质就是：矩阵乘法 + 激活函数

不断重复：

`XW + b → activation`

形成多层网络。
