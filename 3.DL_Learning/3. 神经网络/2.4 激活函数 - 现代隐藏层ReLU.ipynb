## 2.4 激活函数 - ReLU

#### 1. 什么是 ReLU 激活函数
ReLU 的全称是：
`Rectified Linear Unit（修正线性单元）`

它是现代深度学习中最重要、最常用的激活函数之一。

数学公式非常简单：
`ReLU(x) = max(0,x)`


##### 1.1 ReLU 的函数形状
```
输入	输出
x < 0	0
x = 0	0
x > 0	x
```
所以：
* 左侧全部为 0
* 右侧是 线性增长

##### 1.2  ReLU 的输出范围
ReLU 的输出范围是：
`[0 , +∞)`

即：
* 输出永远 非负
* 没有上限

#### 2. ReLU 的导数
ReLU 的导数非常简单：
```
ReLU'(x) = 1 (x > 0)
ReLU'(x) = 0 (x <=0)
```
在：`x = 0`处不可导，但在实际计算中：

通常直接设为：0 或 1对训练几乎没有影响。

#### 3. 为什么 ReLU 会成为最常用的激活函数
在深度学习发展过程中，人们逐渐发现：
ReLU 在训练深层神经网络时效果明显优于 Sigmoid 和 Tanh。
主要有三个原因。

##### 3.1 解决梯度消失问题
Sigmoid 和 Tanh 在输入较大或较小时，会进入 饱和区：
```
σ(x) ≈ 1
σ(x) ≈ 0
```

导数：
`≈ 0`

当网络很深时：
`小梯度 × 小梯度 × 小梯度 → 0`

这就是：**Vanishing Gradient（梯度消失）**

而 ReLU 的导数是：
```
ReLU'(x) = 1   (x > 0)
ReLU'(x) = 0   (x < 0)
```

也就是说：
```
只要输入大于 0：
梯度 = 1
```
不会逐渐变小。

因此：**⭐ ReLU 可以有效缓解梯度消失问题。**

##### 3.2 计算速度非常快
Sigmoid / Tanh 需要计算：`e^x`指数运算。

而 ReLU 只需要：`max(0,x)`,这只是一次简单比较运算。

在大规模神经网络中：
* 数百万
* 数千万
* 甚至数十亿次计算

这种差异会非常明显。

因此：**⭐ ReLU 的计算效率非常高.**

##### 3.3 稀疏激活（Sparse Activation）
ReLU 的一个特点是：`x < 0 → 输出 = 0`

这意味着：**很多神经元会被 关闭。**

例如：
```
[-3 , -1 , 0.5 , 2]
↓
[0 , 0 , 0.5 , 2]
```
很多值变为：0

这就叫：**稀疏激活（Sparse Activation）**

优点：
* 网络更简单
* 计算效率更高
* 减少过拟合

#### 4. ReLU 的缺点
虽然 ReLU 非常优秀，但它也存在一个著名的问题。

##### 4.1 Dead ReLU（死亡 ReLU）
如果神经元的输入一直是：
`x < 0`
那么输出永远是：
`0`
并且：
`梯度 = 0`

这意味着：
`这个神经元 永远不会再更新参数。`

这就叫：**Dead ReLU（死亡 ReLU）**

例如：
* 输入 = -2
* ReLU(-2) = 0
* 梯度 = 0

如果这种情况一直发生：**这个神经元就 彻底失效。**

#### 5. ReLU 在神经网络中的应用
```
网络位置	是否使用 ReLU
隐藏层	非常常用
输出层	很少
CNN	广泛使用
Transformer	通常使用变体
```

例如：
在 CNN 中：
`Conv → ReLU → Conv → ReLU → Pooling`

#### 6. ReLU 的改进版本
为了改善 Dead ReLU 问题，研究人员提出了一些改进版本。

##### 6.1 Leaky ReLU
```
f(x) = x (x>0)
f(x) = ax (x<0)
```

其中：`α ≈ 0.01`

也就是说：
`负区间不再是 0`

而是：
`0.01x`

这样神经元不会完全“死亡”。

##### 6.2  Parametric ReLU (PReLU)
与 Leaky ReLU 类似，但：
`α 是可学习参数`

模型可以自动学习最合适的负斜率。
