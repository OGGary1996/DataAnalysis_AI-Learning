## 6.1 神经网络参数优化 - 梯度下降算法回顾

#### 1. 为什么需要参数优化
在神经网络训练过程中，我们的目标是：

`找到一组参数，使损失函数（Loss Function）最小。`

神经网络中的参数主要包括：
* 权重 W
* 偏置 b

训练的本质就是：

`不断调整 W 和 b，使得模型预测值与真实值的误差最小。`

#### 2. 梯度下降的核心思想

##### 2.1 什么是梯度（Gradient）
梯度表示：

`函数增长最快的方向`

如果我们想 最小化损失函数，就应该：

`沿着梯度的反方向移动`

原因：
* 梯度方向 → 函数增加最快
* 梯度反方向 → 函数下降最快

因此：

`梯度下降 = 沿着梯度反方向更新参数`

##### 2.2 参数更新公式
梯度下降的基本更新公式：
`w新 = w旧 - 学习率*梯度`

我们需要寻找梯度的**反方向**，所以用负号

表示：
* 在初始参数指定完毕之后，第一轮更新：
* 用初始参数 + 梯度的反方向的w
* 得到新的w

实现的效果是：
* w从很小的初始值
* 不断的往梯度的反方向靠近

##### 2.3 用直觉理解梯度下降
可以把梯度下降想象成：

`在山上找最低点`

假设：
* 山的高度 = Loss
* 当前位置 = 当前参数
* 目标 = 找到最低点

过程：
1. 计算当前坡度（梯度）
2. 判断往哪里走下降最快
3. 走一小步
4. 重复这个过程

最终就会到达山谷。

#### 3. 神经网络中的梯度计算
在神经网络中：

参数很多：
* 第一层权重
* 第二层权重
* 偏置
* …

损失函数：

`L(W_1,W_2,...,b_1,b_2,...)`

因此需要计算梯度，这个过程依赖：

`反向传播（Backpropagation）`

步骤：
1. 前向传播计算预测值
2. 计算 Loss
3. 反向传播计算梯度
4. 使用梯度下降更新参数

#### 4. 梯度下降的三种主要形式
假设训练集一共有 10,000 个样本。

在机器学习和深度学习中，梯度下降主要有三种形式：

##### 4.1 Batch Gradient Descent（批量梯度下降）
每一次更新参数：
* 数据量：一次性把 10,000 个样本全都跑完
* 计算：
    * 计算整个训练集的总 loss，得到的是**所有 N 个样本误差平方和的平均值**
    * 将这个总的Loss，针对所有层的W链式法则求偏导，得到梯度。
* 最后只更新 一次参数
* 影响：得到的梯度是“最精准”的，它严格指向损失函数全局下降最快的方向。

`1个 epoch，batch_size=10,000`

跑完整个训练集 → 只更新1次参数

优点 👍
* 梯度稳定
* 收敛方向准确

缺点 👎
* 计算成本非常大
* 数据集大时非常慢

##### 4.2 Stochastic Gradient Descent（随机梯度下降）
每次更新：
* 数据量：只随机抽取1个样本。
* 计算：
    * 算 1 个样本的 loss
    * 用1个Loss，针对所有层的W链式法则求偏到，得到梯度
* 最后更新N次参数
* 影响：你只根据一个人的“意见”来决定前进方向。这个方向虽然含有大量噪声（可能走弯路），但计算极快，且有助于跳出局部最优解。

```
1个 epoch, batch_size=1
10000个样本 → 更新10000次参数
```

优点 👍
* 计算速度快
* 可以在线学习

缺点 👎
* 梯度波动大
* 收敛不稳定

##### 4.3 Mini-Batch Gradient Descent（小批量梯度下降）
✅ 深度学习中最常用的方法。

在我们平时训练神经网络时，通常流程是：
1. 取一个 batch
2. 用这个 batch 做前向传播
3. 得到这个 batch 的 loss
4. 反向传播
5. 计算的梯度也是 基于这个 batch
6. 更新参数

假设 batch size = 100

一次更新参数时：
* 数据量：每次取 batch size=100 个样本。
* 计算：
    * 计算这 100 个样本的 batch loss，得到的是**平均Loss**
    * 将这个 batch Loss，针对所有层的W链式法则求偏导，得到梯度。
* 最后更新100次参数
* 影响：这是折中方案。它比单个样本更稳定，又比全量数据更高效。同时，它能利用现代 CPU/GPU 的并行计算（矩阵运算）优势。

```
1个 epoch:
10000个样本 ÷ 100
= 100个batch
= 更新100次参数
```
优点 👍
* 计算效率高
* 梯度稳定
* GPU 计算友好

#### 5. 神经网络训练完整流程
Mini-batch 训练流程：

```
1. 前向传播
   对一个 batch 的数据进行前向传播，
   每个样本会产生一个 loss，
   然后对 batch 中所有 loss 求平均，
   得到 batch loss（一个标量）。

2. 反向传播
   以 batch loss 为起点，
   利用链式法则进行反向传播，
   逐层计算每一层参数的梯度：

   ∂L/∂W1
   ∂L/∂W2
   ...
   ∂L/∂WL

3. 参数更新
   使用梯度下降更新每一层参数：

   W_l = W_l - η * ∂L/∂W_l
```

#### 6. 梯度下降面临的问题
在神经网络训练中，简单的梯度下降存在一些问题：

**1️⃣ 学习率难以选择**

如果学习率：
* 太大
    * 可能震荡甚至发散
* 太小：
    * 收敛非常慢

**2️⃣ 容易陷入局部最优**

损失函数通常不是简单曲线，而是复杂曲面：
```
局部最优
鞍点
平坦区域
```

**3️⃣ 收敛速度慢**

普通梯度下降：下降速度较慢

特别是在深度网络中。


#### 7. 为什么需要优化算法
为了解决梯度下降的问题，深度学习提出了许多 **优化算法**：

常见优化算法：
```
优化算法	特点
Momentum	增加惯性
AdaGrad	自适应学习率
RMSProp	改进AdaGrad
Adam	最常用优化算法
```
这些算法本质上：

都是在改进梯度下降
