## 2.2 激活函数 - Softmax & CrossEntropy

#### 1. 什么是 Softmax 激活函数
Softmax 是神经网络中非常重要的输出层激活函数，主要用于：**⭐ 多分类问题（Multi-class Classification）**

例如：
* 手写数字识别（0~9）
* 图像分类（猫、狗、鸟）
* 医疗影像分类（不同疾病类型）

Softmax 的作用是：把一组任意实数转换成一个概率分布。

也就是说：
* 所有输出值都在 0~1之间
* 所有输出加起来 等于1 (概率)

#### 2. Softmax 的数学公式
假设神经网络输出一组值：
`z = [z_1, z_2, z_3, ..., z_n]`

Softmax 的计算公式是：`Softmax(z_i) = e^z_i / SUM(e^z_i)`

解释：
```
符号	含义
z_i	第 i 个类别的原始得分
e^{z_i}	对得分进行指数放大
分母	所有类别指数的总和
```

其中分母部分是所有的指数的和，相当于把每一个指数进行**归一化**

最终得到：P_i, 表示第i个类别的概率

#### 3. Softmax 的计算过程（直观理解）
假设神经网络输出：
`z = [2.0 , 1.0 , 0.1]`

这三个值叫做：
**⭐ logits（未归一化得分）**

##### 3.1 步骤 1：计算指数
```
e^2.0 ≈ 7.39
e^1.0 ≈ 2.72
e^0.1 ≈ 1.11
```

##### 3.2 步骤 2：计算总和
`7.39 + 2.72 + 1.11 = 11.22`

##### 3.3 步骤 3：归一化
```
7.39 / 11.22 = 0.66
2.72 / 11.22 = 0.24
1.11 / 11.22 = 0.10
```

##### 3.4 最终结果：
`Softmax(z) = [0.66 , 0.24 , 0.10]`

可以解释为：
```
类别	概率
类别1	66%
类别2	24%
类别3	10%
```
所有概率加起来：
`0.66 + 0.24 + 0.10 = 1`

#### 4. Softmax 的作用
Softmax 的核心作用：**⭐ 把模型输出转换为概率分布**

这样可以直接解释为：
`P(class | x)`

例如：
`[0.05 , 0.10 , 0.80 , 0.05]`

表示：
```
类别	概率
class1	5%
class2	10%
class3	80%
class4	5%
```

模型预测：`class3`因为概率最大。

#### 5. Softmax 与 Sigmoid 的区别
Sigmoid - ⭐ 二分类问题

输出：
`P(y=1)`

例如：
`0.8`

表示：
`80% 概率是正类`

Softmax - ⭐ 多分类问题
输出：
```
P(class1)
P(class2)
P(class3)
...
```

所有概率加起来：
`= 1`


#### 6. Softmax 的重要特点

##### 6.1 概率归一化
Softmax 的最大特点：
`∑ P_i = 1`

因此输出可以直接解释为概率。

##### 6.2 放大差异
因为 Softmax 使用指数：`e^x`

指数函数会：⭐ 放大大的数值

例如：
`z = [5 , 1]`

计算：
```
e^5 ≈ 148
e^1 ≈ 2.7
```

Softmax 结果：
`≈ [0.98 , 0.02]`

差异被明显放大。

##### 6.3 输出互相影响
Softmax 的输出是：互相关联的

因为：分母是所有类别的和

如果一个类别概率增加：

其他类别概率就会减少。

#### 7. Softmax 与 CrossEntropy交叉熵损失函数
在深度学习中：

**⭐ Softmax 几乎总是与 CrossEntropy 一起使用**

训练流程：
```
logits
↓
Softmax
↓
CrossEntropy Loss
```

例如：

真实标签：
`[0 , 0 , 1]`

预测概率：
`[0.1 , 0.2 , 0.7]`

CrossEntropy 会计算：
`预测概率 与 真实概率 的差距`

并作为损失函数。

#### 8. 重要实践知识（非常关键）
在 PyTorch 中训练分类模型时：

**⚠️ 通常不需要手动写 Softmax**

例如：
`nn.CrossEntropyLoss()`

这个损失函数内部已经包含：
`Softmax + Log`

因此训练代码通常是：
`logits → CrossEntropyLoss`

而不是：
`Softmax → CrossEntropyLoss`

否则会影响数值稳定性。
