## 1.1 神经网络基础 - 神经网络的结构

#### 1. 基本组成部分：

##### 1.1 神经网络由哪些部分组成？：
最基本的神经网络通常包含三部分：
输入层（Input Layer） → 隐藏层（Hidden Layer） → 输出层（Output Layer）

```
输入层        隐藏层        输出层
x1   ───┐
x2   ───┼──►  ○  ○  ○ ───► y
x3   ───┘
```

解释：
* 输入层：接收原始数据
* 隐藏层：对数据进行加工、提取规律
* 输出层：给出最终结果

例如房价预测：
* 输入层：面积、地段、房龄、楼层
* 隐藏层：学习这些因素之间复杂的组合关系
* 输出层：预测房价

例如手写数字识别：
* 输入层：图片像素值
* 隐藏层：学习笔画、结构、形状
* 输出层：输出是 0~9 中哪个数字


##### 1.2 什么叫“隐藏层”？
隐藏层之所以叫“隐藏”，不是因为它神秘，而是因为：
* 它不直接接收现实世界原始数据
* 也不直接输出最终答案
* 它位于中间，负责内部计算

也就是说，隐藏层的作用就是：
把原始输入逐步变成更有意义、更适合任务的表示

##### 1.3  什么是“深度”？
如果神经网络只有 1 个隐藏层，那就是比较浅的网络。

如果有很多隐藏层，那就是深度神经网络（Deep Neural Network）。

所以：
* 神经网络：广义概念
* 深度学习：使用很多层的神经网络进行学习

“深度”说的其实就是层数更深。

#### 2. 一个神经元到底在做什么？

##### 2.1 一个神经元的计算过程
一个神经元接收多个输入，例如：
* x1
* x2
* x3

每个输入都有一个对应的重要程度，这个重要程度叫做权重（weight）：
* w1
* w2
* w3

然后神经元先做一步：加权求和
`z = w1x1 + w2x2 + w3x3 + b`

其中：
* w：权重
* x：输入
* b：偏置（bias）
* z：线性计算结果

然后再经过一个激活函数（activation function）：
`a = f(z)`
最终输出 a，再传给下一层。


##### 2.2 为什么要有权重？
权重可以理解成：模型认为每个输入有多重要

例如判断一个房子的价格：
* 面积可能很重要
* 地段可能更重要
* 房龄可能有影响但稍弱

那么不同输入前面的权重就会不同。

在训练过程中，模型会自动调整这些权重。

所以，神经网络学习的过程，本质上很大程度上就是：

不断调整权重和偏置，让输出更接近真实答案

##### 2.3 为什么要有偏置 bias？
偏置可以理解为一个“额外的可调整量”，它让模型更灵活。

如果只有：
`z = w1x1 + w2x2`
那么这个关系会受到一定限制。

加上 bias 后：
`z = w1x1 + w2x2 + b`
模型就可以整体平移，更容易拟合数据。

你可以把 bias 理解成：
* 在加权求和基础上的一个修正项
* 它和线性回归中的截距非常像。

##### 2.4 为什么一个神经元不够？
因为一个神经元本质上只能做比较有限的表达。
多个神经元组合起来，才能学习更复杂的规律。

可以把它理解成一个团队：
* 一个神经元负责感知某种模式
* 另一个负责感知另一种模式
* 多个神经元组合起来，最终完成复杂判断

所以神经网络不是靠一个神经元厉害，而是靠大量神经元协同工作。

#### 3. 最重要也最难理解的部分：为什么一定要有激活函数？
这是神经网络里的超级核心知识点。⭐

##### 3.1 如果没有激活函数会怎样？
假设每一层都只是做线性计算：
`z = Wx + b`

那看起来有很多层，好像很复杂，但本质上：
多层线性变换叠加后，结果仍然是线性变换

也就是说：
* 第一层线性
* 第二层线性
* 第三层线性

最后整体还是一个“大一点的线性模型”。

这意味着，即使你堆很多层，没有激活函数，模型依然学不了真正复杂的非线性关系。

所以：
激活函数的作用，就是给神经网络加入非线性能力。

##### 3.2 什么叫“非线性能力”？
线性模型只能学比较简单的关系，例如一条直线、一个平面。

但现实世界大多数问题都不是线性的：
* 图片分类不是线性的
* 语言理解不是线性的
* 医学诊断不是线性的
* 语音识别不是线性的

所以模型必须能够表示“弯曲的、复杂的、多层次的关系”，这就需要非线性。

##### 3.3  激活函数到底做了什么？
激活函数接收线性结果 z，再做一次变换：
`a = f(z)`

这个函数通常不是简单的直线函数，因此它能让每个神经元的输出具有非线性特征。

这样一来，整个网络就不再只是“线性叠加机器”，而变成了能够学习复杂模式的模型。
