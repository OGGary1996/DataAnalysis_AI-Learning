## 3.3 MLP 基础ANN - 参数初始化的必要性

#### 1. 什么是参数初始化？

##### 1.1 参数初始化的含义
在神经网络中，参数主要指两类：
* 权重（weight）
* 偏置（bias）

在模型训练开始之前，这些参数不可能是空的，必须先给它们一个初始值。

这个“在训练前先给参数赋值”的过程，就叫做：**参数初始化（Parameter Initialization）**

##### 1.2 为什么训练前一定要先初始化？
因为神经网络的训练，本质上是：

从一组初始参数出发，不断通过梯度下降去调整参数，让损失逐渐变小。

所以训练流程其实是：
* 先有一组初始参数
* 再进行前向传播
* 计算损失
* 反向传播求梯度
* 更新参数

如果一开始参数没有值，后面的计算就根本无法进行。

##### 1.3 参数初始化不是“随便给个数”
很多初学者会觉得：反正后面都会训练，那一开始随便设个值不就行了？

其实不是。⚠️

参数初始化会直接影响：
* 模型能不能顺利开始训练
* 梯度会不会太大或太小
* 收敛速度快不快
* 训练是否稳定
* 深层网络能不能学到东西

所以参数初始化虽然发生在训练最开始，但它对整个训练过程影响非常大。

#### 2. 参数初始化到底在初始化什么？

##### 2.1 权重初始化
权重是神经网络最核心的可学习参数。

例如一层线性层：
* 输入特征数：in_features
* 输出神经元数：out_features

在 PyTorch 中，权重矩阵通常形状为：

`W.shape = (out_features, in_features)`

例如：
* 输入 3 个特征
* 输出 4 个神经元

那么：

`W.shape = (4, 3)`

这些权重在训练前都要先初始化。

##### 2.2 偏置初始化
偏置也是可学习参数。

在 PyTorch 中，bias 通常是一维向量：

`b.shape = (out_features,)`

例如输出层有 4 个神经元：

`b.shape = (4,)`

一般来说：
* 权重初始化更关键
* 偏置通常初始化为 0 或很小的常数

因为训练中的主要表达能力，更多来自权重矩阵。

#### 3. 为什么不能把权重全部初始化为 0？
这是参数初始化中最重要的第一条知识点。⭐

##### 3.1 直觉上看，好像全 0 很整齐
* 所有权重都设成 0
* 所有偏置也设成 0

这样看起来很“公平”，但实际上这是错误的。

##### 3.2 核心问题：会造成对称性（symmetry）
假设某一层有多个神经元，如果它们的初始权重完全一样，例如都为 0，那么：
* 它们前向传播时输出相同
* 它们计算出的损失影响相同
* 反向传播得到的梯度也相同
* 参数更新后仍然相同

本来应该学习不同特征的多个神经元，最后全都学成了一模一样的东西。

这就失去了“多个神经元”的意义。

##### 3.3 为什么这会很严重？
因为神经网络希望：
* 有的神经元学边缘
* 有的神经元学纹理
* 有的神经元学局部模式
* 有的神经元学组合特征

但如果大家从一开始就完全一样，训练过程中也一直一样，那么这一层虽然看起来有很多神经元，实际上就像只有 1 个神经元 在工作。😵

##### 3.4 结论
✅ 权重不能全部初始化为相同值，特别不能全部初始化为 0。

但要注意：

✅ bias 初始化为 0 通常是可以的。

因为真正决定“不同神经元是否不同”的关键，是权重，而不是偏置。

#### 4. 参数初始化不合理会带来什么问题？

##### 4.1 梯度消失（Vanishing Gradient）
如果参数初始化不合适，尤其是层数很多时，前向传播和反向传播的数值会不断缩小，导致：
* 越靠前的层梯度越接近 0
* 参数几乎不更新
* 模型学不动

##### 4.2 梯度爆炸（Exploding Gradient）
如果初始化值过大，层与层之间的信号可能不断放大，导致：
* 激活值越来越大
* 梯度越来越大
* 参数更新过猛
* loss 震荡甚至直接变成 NaN

##### 4.4 收敛很慢
即使没有明显梯度消失或爆炸，如果初始化不好，也可能导致：
* loss 下降很慢
* 训练前期效率低
* 模型需要更多 epoch 才能进入有效学习状态

##### 4.5 模型难以训练深层网络
浅层网络有时还能勉强学，但一旦网络层数加深，不合理的初始化问题会被放大很多倍。
