#### 2. 完整流程

##### 2.1 搭建MLP模型

````python
import torch
import torch.nn as nn

class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer_1 = nn.Linear(3, 4)
        self.relu = nn.ReLU()
        self.layer_2 = nn.Linear(4, 1)
    
    def forward(self, x):
        x = self.layer_1(x)
        x = self.relu(x)
        x = self.layer_2(x)
        return x
````

##### 2.2 训练模型

````python
X = torch.randn(10,3) # 10个样本，每个样本3个特征
y = torch.randn(10,1) # 10个样本，每个样本1个目标值

# 创建模型对象
model = MLP()

# 定义损失函数，这里采用回归的均方误差损失
loss = nn.MSELoss()

# 定义优化器，这里采用随机梯度下降
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# 训练模型：
for epoch in range(100):
    # forward pass
    y_pred = model(X)
    # 计算损失
    l = loss(y_pred, y)
    # backward pass
    l.backward()
    # 更新权重
    optimizer.step()
    # 清零梯度
    optimizer.zero_grad()
    if epoch % 10 == 0:
        print(f'Epoch {epoch}, Loss: {l.item()}')
````

#### 3. 训练过程详解

##### 3.1 定义损失函数（Loss Function）
损失函数用于衡量：

`模型预测结果和真实标签之间的差距。`

例如：
* MSELoss → 回归
* CrossEntropyLoss → 分类
    * 交叉熵损失函数
    * 与 SoftMax 搭配使用

##### 3.2 定义优化器（Optimizer）
优化器负责：

`根据梯度更新模型参数。`

常见优化器：
* SGD
* Adam


##### 3.3 前向传播（Forward Pass）
把输入数据送入模型：

`y_pred = model(X)`

得到预测结果。

##### 3.4 计算损失
用损失函数计算：

`loss = criterion(y_pred, y_true)`

##### 3.5 反向传播（Backward Pass）
计算梯度：

`loss.backward()`

##### 3.6 更新参数
使用优化器更新参数：

`optimizer.step()`

##### 3.7 清空梯度
PyTorch 默认会累积梯度，因此需要手动清空：

`optimizer.zero_grad()`
