## 3 RNN基础介绍

#### 1、RNN 是什么 🧠

##### 1.1 RNN 的全称

RNN 的英文全称是 **Recurrent Neural Network**，中文一般翻译为：

**循环神经网络**

这里的“循环”不是说网络结构画成了一个圈，而是指：

👉 网络在处理当前数据时，会把前面时刻的信息也一起带入当前计算中。

也就是说，RNN 和我们之前学习的 MLP、CNN 最大的不同就在于：

- **MLP / CNN：** 更擅长处理“当前这一份输入”
- **RNN：** 更擅长处理“有先后顺序的数据”

##### 1.2 什么叫“有顺序的数据”

有些数据不是孤立存在的，而是 **前后有关联** 的。

例如：

- 一句话中的单词，是一个接一个出现的 📖
- 一段语音中的声音，是一帧一帧变化的 🎧
- 每天的温度，是按时间连续记录的 🌡️
- 股票价格、销量、传感器数据，都是时间序列 📈

这些数据有一个共同特点：

**当前的信息，往往和前面的信息有关。**

比如一句话：

> “I am very …”

如果后面接的是：

- happy
- sad
- tired

前面的单词会影响我们对后面单词的理解。

所以这类任务中，模型不能只看当前这一个输入，而要能够：

- ✅ 记住前面的内容
- ✅ 结合上下文来理解当前内容

这就是 RNN 要解决的问题。

##### 1.3 RNN 的核心思想

RNN 的核心思想可以用一句话概括：

**当前时刻的输出，不仅取决于当前输入，还取决于之前时刻留下来的“记忆”。**

这个“记忆”，在 RNN 中通常叫做：

**hidden state（隐藏状态）**

你可以把它理解成：

🧠 模型脑子里暂时记住的信息

当新的输入到来时，RNN 会做两件事：

- 接收当前输入
- 结合上一时刻的隐藏状态，更新新的隐藏状态

所以，RNN 天然适合处理序列。

#### 2、为什么需要 RNN ❓

##### 2.1 普通神经网络为什么不够用

我们之前学过的 MLP，本质上是：

- 输入一组特征
- 经过若干层计算
- 输出一个结果

它默认假设：

**每一条输入样本是独立的。**

例如房价预测：

- 面积
- 房间数
- 地段

这些特征一起输入，模型直接预测价格。

这个过程没有“前后顺序”的概念。

##### 2.2 CNN 为什么也不能很好解决序列记忆问题

CNN 非常适合处理图像，因为它擅长：

- 提取局部特征
- 识别空间结构
- 利用卷积核发现模式

比如图像中：

- 边缘
- 纹理
- 角点
- 更复杂的形状

CNN 的重点是：

📌 **空间上的局部关联**

但是对于很多序列任务，真正重要的是：

📌 **时间上的前后依赖关系**

例如一句话：

> “I did not like this movie.”

如果只看后面的 “like” 和 “movie”，你可能觉得是正向情感；  
但前面的 **not** 会改变整个句子的意思。这说明：

**序列任务中，前面的内容会影响后面的理解。**

CNN 虽然也可以通过一维卷积处理序列，但它本质上更偏向于提取局部模式，  
而 RNN 的设计目标更直接：

- ✅ 专门处理“前后相关”的问题
- ✅ 让模型拥有一定的“记忆能力”

##### 1.3 一个最直观的例子

我们来看一句话：

> “The food in this restaurant is not good.”

如果模型逐个单词处理：

`The -> food -> in -> this -> restaurant -> is -> not -> good`

当模型读到最后的 **good** 时，不能只看这个单词本身。

因为真正的意思要结合前面的 **not**。

所以在这里，模型必须具备：

- 记住前面内容的能力
- 把历史信息传递到当前时刻的能力

这正是 RNN 的价值所在。


#### 3、RNN 适合处理什么问题 📌

RNN 最适合处理的，是 **序列型数据（Sequential Data）**。

##### 3.1 自然语言处理（NLP）

这是 RNN 最经典的应用场景之一。

例如：

- 文本分类
- 情感分析
- 机器翻译（早期经典方法）
- 词性标注
- 命名实体识别
- 自动文本生成

为什么适合？

因为文本天然就是按顺序排列的：

- 单词有顺序
- 句子有上下文
- 前文会影响后文

##### 3.2 时间序列预测

例如：

- 气温预测
- 股票价格预测
- 销量预测
- 用电量预测
- 传感器数据预测

这些数据是随着时间不断变化的，  
过去的值往往会影响未来的值。

例如：

- 前 7 天的销量 → 预测第 8 天销量
- 前 24 小时的温度 → 预测下一小时温度

这类任务中，顺序很重要，RNN 非常适合入门理解。

##### 3.3 视频或动作序列分析

视频本质上是一帧一帧组成的时间序列。

例如：

- 动作识别
- 行为分析
- 手势识别

虽然现代方法中 CNN + Transformer 更常见，  
但从原理学习角度，RNN 很适合理解“时间上的连续变化”。

#### 4、RNN 和 MLP、CNN 的本质区别 🔍

##### 4.1 MLP：处理独立样本

MLP 更像是：

> 给你一组固定特征，你直接输出结果

它不关心顺序，也不保留历史状态。

##### 4.2 CNN：处理空间局部模式

CNN 更像是：

> 在局部区域中找特征，再逐层组合成更复杂的结构

它最擅长的是：

- 图像中的空间关系
- 网格数据中的局部特征提取

##### 4.3 RNN：处理前后依赖关系

RNN 更像是：

> 一边读，一边记；读到后面时，脑子里还保留前面的信息

所以三者可以简单对比为：

- **MLP：** 不考虑顺序
- **CNN：** 强调空间局部结构
- **RNN：** 强调时间/顺序依赖与上下文记忆


#### 5、如何理解 RNN 中的“记忆” 🧠

##### 5.1 生活中的类比

你可以把 RNN 想象成一个人在读句子。

比如读：

> I am learning deep learning today.

人类不会每看到一个单词，就把前面的全忘掉。  
而是会把前面的内容暂时记在脑中，再继续往后读。

例如：

- 读到 I
- 再读到 am
- 再读到 learning
- 再读到 deep learning
- 最后读到 today

整个理解过程是连续的。

RNN 也是类似的逻辑：

- 每接收一个输入
- 更新一次内部记忆
- 把新的记忆传给下一步

##### 5.2 隐藏状态 hidden state

RNN 中最重要的概念之一就是：

**隐藏状态（hidden state）**

它可以理解为：

📦 **到当前时刻为止，模型总结出来的一份“上下文信息包”**

这个信息包里装着什么？

- 可能是前面单词的语义线索
- 可能是前面时间点的变化趋势
- 可能是前面若干步的重要特征

**hidden state 就是 RNN 的短期记忆。**

#### 6、RNN 的基本处理方式 ⏱️

##### 6.1 按时间步一步一步处理

RNN 和普通网络不一样，它不是“一次性看完整个序列”，而是：

**按顺序一个一个地处理输入。**

例如一个序列：

`x_1, x_2, x_3, x_4`

RNN 的处理方式是：

- 第一步读入 `x_1`
- 第二步读入 `x_2`，同时结合前一步记忆
- 第三步读入 `x_3`，继续更新记忆
- 第四步读入 `x_4`，得到新的状态或输出

所以 RNN 的本质是：

- ✅ 序列递推
- ✅ 状态传递
- ✅ 参数重复使用

##### 6.2 为什么这叫“循环”

因为在结构图中，当前时刻的隐藏状态会传递到下一时刻，  
看起来像是有一条“回路”。

但实际计算时，我们通常会把它展开成一条时间链。

后面你会学到：

- 循环结构图
- 展开图（unrolled RNN）

这一步是理解 RNN 的关键。


#### 7、RNN 的优点与局限 ⚖️

##### 7.1 优点

RNN 的优点主要有：

- 能处理变长序列
- 能利用前后顺序信息
- 能保留一定的上下文记忆
- 非常适合作为序列模型的入门结构

例如：

- 一个句子可以长一点，也可以短一点
- 一段时间序列可以有 10 步，也可以有 100 步

RNN 都可以逐步处理。

##### 7.2 局限

虽然 RNN 很重要，但它也有明显问题：

###### （1）记忆能力有限

序列太长时，早期信息可能很难一直保留下来。

例如一句很长的话，  
模型读到最后时，可能已经“忘了”开头的重要内容。

###### （2）训练时容易出现梯度消失或梯度爆炸

这部分我们后面会详细学习。

👉 **传统 RNN 在长序列任务中训练比较困难。**

###### （3）难以处理长期依赖

所谓长期依赖，就是：

**当前结果依赖很久以前的信息**

例如：

> “I grew up in France … I speak fluent French.”

如果两个关键信息中间隔了很多词，  
传统 RNN 往往不容易稳定记住前面的内容。

也正因为如此，后来才发展出了：

- **LSTM**
- **GRU**

它们是对普通 RNN 的改进版本。

#### 8、为什么我们仍然要先学 RNN 🌱

你可能会问：

既然 RNN 有局限，为什么不直接学 LSTM 或 Transformer？

原因是：

##### 8.1 RNN 是序列模型的起点

RNN 是最基础的序列神经网络。

如果不理解 RNN，就很难真正理解：

- LSTM 为什么要加门控
- GRU 为什么要简化结构
- Transformer 为什么敢彻底去掉循环

##### 8.2 RNN 能帮助我们建立“序列建模”的直觉

学习 RNN，最重要的不是它现在是不是最先进，  
而是它能帮助你理解：

- 什么是序列输入
- 什么是上下文
- 什么是状态传递
- 什么是长期依赖问题

这些思想在后续学习中都非常重要。

##### 8.3 它是理解后续模型的桥梁

后面的学习路线通常是：

`Simple RNN → LSTM / GRU → Attention → Transformer`

所以 RNN 是必经的一步。
