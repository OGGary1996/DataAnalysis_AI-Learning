## 5.1 Pytorch 模拟线性回归 - 数据准备

#### 1. 我们要解决什么问题？

我们想模拟一个最经典的线性关系：`y = 3x + 2`

但真实世界的数据通常是：`y = 3x + 2 + 噪声`

所以我们要做的事情是：
1. 构建一组x
2. 根据公式生成y
3. 添加随机噪声
4. 将x转换为符合模型输入条件的 tensor

#### 2. 手动构造简单数据

##### 2.1 构建x

```python
import torch

# 构建一组简单的数据
x = torch.arange(1,11, dtype=torch.float32)
print(x)
```

##### 2.2 构建y 并且添加随机噪声:
1. torch.randn() 生成标准正态分布随机数 
2. 乘以0.5控制噪声强度

```python
true_w = 3
true_b = 2

noise = torch.randn(x.shape) * 0.5

y = true_w * x + true_b + noise
print(y)
```

#### 3. x 形状调整：

在 PyTorch 的线性层中，输入必须是：`(batch_size, feature_dim)`
* batch_size：表示样本数量
* feature_dim：表示特征维度大小

而 `x.shape = (10,)` 不满足要求，需要通过 `reshape()` 转换为2D

```python
x = x.reshape(-1,1) # 使用 -1 来自动推断行数
print(x.shape)
print(y.shape)
```
