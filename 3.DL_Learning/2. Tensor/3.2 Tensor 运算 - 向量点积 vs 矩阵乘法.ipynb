## 3.2 Tensor 计算 - 向量点积 vs 矩阵乘法

#### 1. 逐元素乘法：
逐元素乘法是指两个张量的对应元素相乘，得到一个新的张量。<br>
对于两个**形状相同**的张量A和B，逐元素乘法可以表示为：
C = A * B
其中，C的每个元素C[i][j]等于A[i][j]乘以B[i][j]。逐元素乘法也称为Hadamard积。

```python
import torch

a = torch.tensor([[1, 2],
                  [3, 4]])
b = torch.tensor([[5, 6],
                  [7, 8]])

print("逐元素乘法结果：")
print(a * b)
```

#### 2. 向量点积：
向量点积是指两个1维向量(1维张量)的对应元素相乘后求和，得到一个标量。<br>
对于两个**一维**张量A和B，向量点积可以表示为：
C = A · B = A[0] * B[0] + A[1] * B[1] + ... + A[n-1] * B[n-1]
其中，C是一个标量，表示A和B的点积。

```python
c = torch.tensor([1, 2, 3])
d = torch.tensor([4, 5, 6])
print("向量点积结果：")
print(torch.dot(c, d))
```

#### 3. 矩阵乘法：

##### 3.1 规则：
无论多少维，矩阵乘法（Dot Product / Matmul）必须满足：<br>
左矩阵的最后一维 = 右矩阵的倒数第二维 <br>
比如：
- 二维张量：A（M x N）和 B（N x P），结果是C（M x P）
- 三维张量：A（L x M x N）和 B（L x N x P），结果是C（L x M x P）

##### 3.2 示例：二维矩阵乘法：
- 张量 A (形状 2×3): 2行，3列
- 张量 B (形状 3×2): 3行，2列 <br>
A = [[1, 2, 3],
     [4, 5, 6]]
B = [[7, 8],
     [9, 10],
     [11, 12]]
计算过程：
1. C11（A第一行，B第一列）的点积：1x7 + 2x9 + 3x11 = 58
2. C12（A第一行，B第二列）的点积：1x8 + 2x10 + 3x12 = 64
3. C21（A第二行，B第一列）的点积：4x7 + 5x9 + 6x11 = 139
4. C22（A第二行，B第二列）的点积：4x8 + 5x10 + 6x12 = 154 <br>
结果矩阵C为：
C = [[58, 64],
     [139, 154]]


##### 3.3 示例：三维矩阵乘法：
在深度学习（如处理 MRI 图像数据或 Transformer 模型）中，我们经常遇到三维张量。通常第一维是 Batch Size (批大小)。
- 张量 A (形状 2×2×3): 2个样本，每个样本是一个2行3列的矩阵
- 张量 B (形状 2×3×2): 2个样本，每个样本是一个3行2列的矩阵 <br>
计算过程：
1. 第一组 (Batch 0): 拿 A[0] 与 B[0] 做矩阵乘法，得到 C[0]；结果同上，C[0] = [[58, 64], [139, 154]]。
2. 第二组 (Batch 1): 拿 A[1] 与 B[1] 做矩阵乘法，得到 C[1]；结果同上，C[1] = [[58, 64], [139, 154]]。

#### 4. PyTorch 实现：
- 逐元素乘法：使用 `*` 运算符或 `torch.mul()`
- 向量点积：使用 `torch.dot()`
- 矩阵乘法：使用 `torch.matmul()` 或 `@` 运算符号

#### 5. 1维张量的特殊情况：
对于1维张量，`torch.matmul()` 会执行**向量点积**，而不是逐元素乘法。这是因为1维张量被视为向量，而不是矩阵。

```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])
print("使用 torch.matmul 计算1维张量的结果：")
print(torch.matmul(a, b))  # 输出 32，即向量点积结果
```
