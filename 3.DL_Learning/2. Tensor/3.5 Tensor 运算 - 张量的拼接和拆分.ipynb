## 3.5 Tensor 运算 - 张量的拼接和拆分

#### 1. 拼接操作的本质：
* 拼接操作本质上是将多个张量**按照某个维度**拼接起来

#### 2. cat() 

核心规则：
1. 除了拼接维度以外，其它维度必须完全相同
2. 比如按行拼接，则列必须维度大小相同

```python
import torch

a = torch.randn(2,3)
b = torch.randn(3,3)
# 按照行拼接，dim = 0，此时a和b的列数必须相同
cated_row = torch.cat([a,b], dim=0)
print("Cat along rows (dim=0):", cated_row.shape)
```

```python
a = torch.randn(2,3)
b = torch.randn(2,4)
# 按照列拼接，dim = 1，此时a和b的行数必须相同
cated_col = torch.cat([a,b], dim=1)
print("Cat along columns (dim=1):", cated_col.shape)
```

#### 3. stack

stack的本质：
1. 首先对选择的维度位置增加一个维度，
2. 然后在新增的维度进行堆叠

stack的计算逻辑：可以理解为2步操作 <br>
执行 torch.stack([A, B], dim=i) 等同于：
1. 执行 unsqueeze(dim=i),在i处新增一个大小为1的维度
2. 然后在该维度执行cat操作，进行堆叠

```python
c = torch.randn(3,4)
d = torch.randn(3,4)

# dim = 0，在维度0位置上堆叠
stacked_1 = torch.stack([c,d], dim=0)
print("Stack along dim=0:", stacked_1.shape)

# dim = 1，在维度1位置上堆叠
stacked_2 = torch.stack([c,d], dim=1)
print("Stack along dim=1:", stacked_2.shape)
```

#### 4. 拆分操作：
* 拆分本质上是将张量按某一个维度进行拆分
* 维度数量减少

##### 4.1 chunk 操作 - 指定拆分的分份数

```python
e = torch.randn(4,8)
# 沿着行方向切分成2份，dim = 0
chunck_row = torch.chunk(e, chunks=2, dim=0)
# 得到的chunck_row是一个包含2个张量的元组，每个张量的形状为(2, 8)
for chunck in chunck_row:
    print("After chunking along rows:", chunck.shape)

# 沿着列方向切分成4份，dim = 1
chunck_col = torch.chunk(e, chunks=4, dim=1)
# 得到的chunck_col是一个包含4个张量的元组，每个张量的形状为(4, 2)
for chunck in chunck_col:
    print("After chunking along columns:", chunck.shape)
```

#### 5. split 操作：
split 方法更加灵活，可以：
1. 指定拆分的个数
2. 分别指定每一个部分的大小，不均匀拆分

##### 5.1 指定拆分大小

```python
f = torch.randn(4,8)
# 沿着行方向切分成2份，dim = 0
split_row = torch.split(f, split_size_or_sections=2, dim=0)
# 得到的split_row是一个包含2个张量的元组，每个张量的形状为(2, 8)
for split in split_row:
    print("After splitting along rows:", split.shape)

# 沿着列方向切分成4份，dim = 1
split_col = torch.split(f, split_size_or_sections=2, dim=1)
# 得到的split_col是一个包含4个张量的元组，每个张量的形状为(4, 2)
for split in split_col:
    print("After splitting along columns:", split.shape)
```

##### 5.2 不均匀拆分

```python
# 按照行拆分，第一份包含1行，第二份包含3行
split_row__uneven = torch.split(f, split_size_or_sections=[1,3], dim=0)
for split in split_row__uneven:
    print("After uneven splitting along rows:", split.shape)

# 按照列拆分，第一份包含2列，第二份包含6列
split_col__uneven = torch.split(f, split_size_or_sections=[2,6], dim=1)
for split in split_col__uneven:
    print("After uneven splitting along columns:", split.shape)
```
