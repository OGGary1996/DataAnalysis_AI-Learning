## 1.1 Tensor 的创建方式

#### 1. 张量的基本创建方式

##### 1.1 从python数据创建

```python
import torch
import numpy as np
# 创建0维张量
a = torch.tensor(1)

# 创建1维张量
b = torch.tensor([1, 2, 3])

# 创建2维张量
c = torch.tensor([[1, 2],
                  [3, 4]])

data = np.random.randint(0,10, size=(2,3))
d = torch.tensor(data)

print(a)
print(b)
print(c)
print(d)
```

##### 1.2 指定数据类型创建并且查看数据类型

```python
a = torch.tensor(1, dtype=torch.float32)
print(a.dtype)
```

#### 2. 创建全0张量、全1张量

```python
# 全0
e = torch.zeros((2,3))
print(e)
# 全1
f = torch.ones((2,3))
print(f)
```

#### 3. 创建线性序列张量

##### 3.1 arange 类似于python的range函数和numpy的arange函数

```python
# 从0到10，步长为2
g = torch.arange(0,10, step=2)
print(g)
```

##### 3.2 linspace 在指定范围内生成等差数列, 类似于numpy的linspace函数

```python
# 从0到10，生成5个数
h = torch.linspace(0,10, steps=5)
print(h)
```

#### 4. 创建随机张量

##### 4.1 rand 创建均匀分布的随机数张量（0-1）

```python
i = torch.rand(2,3)
print(i)
```

##### 4.2 randn 创建标准正态分布的随机数张量（均值为0，标准差为1）


```python
j = torch.randn(2,3)
print(j)
```

##### 4.3 randint 创建指定范围内的随机整数张量

```python
k = torch.randint(0,10, size=(2,3))
print(k)
```

#### 5. 元素类型转换

```python
# 将整数张量转换为浮点数张量
k = k.to(torch.float)
print(k)

# 直接方法
k = k.int()
print(k)
```

#### 6. 查看常见属性

```python
# 查看张量的形状
print(k.shape)
# 查看张量的维度
print(k.ndim)
# 查看张量的数据类型
print(k.dtype)
# 查看张量所在的设备
print(k.device)
```
