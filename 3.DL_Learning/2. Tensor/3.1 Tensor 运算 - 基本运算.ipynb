#### 2. 张量与标量的运算：
依赖于广播机制，标量会自动扩展为与张量相同的形状，然后进行逐元素运算。

```python
a = torch.tensor([1,2,3])
print(a * 2) # 每个元素都乘以2
print(a + 5) # 每个元素都加上5
print(a - 1) # 每个元素都减去1
print(a / 2) # 每个元素都除以2

# 等价于将 标量1 扩展为与张量a相同的形状 [1,1,1]，然后进行逐元素运算
```

```python
b = torch.tensor([
    [1,2,3],
    [4,5,6]
])

c = torch.tensor([10,20,30])

print(b + c)
```

#### 3. 广播机制：
1. 当两个张量的形状不同时，PyTorch会自动进行广播，使得它们具有相同的形状，从而可以进行逐元素运算。
2. 广播的规则如下：
  - 比较2个张量的维度的最后一个维度
    - 比如上一个例子中，比较(2,3) 和 (,3)，最后一个维度都是3
  - 如果最后一个维度相等，或者其中一个维度为1，则认为它们是兼容的，可以进行广播

#### 4. 原地操作：
1. 在numpy pandas中，原地操作的参数是 inplace=True
2. 在PyTorch中，原地操作的函数名称以 _ 结尾，比如 add_()、mul_()、div_() 等。
3. 原地操作会直接修改原始张量的值，而不是返回一个新的张量。

```python
data = torch.tensor([1,2,3])
data_add = data.add(10) # 返回一个新的张量，原始张量不变
print("Data:", data)
print("Data + 10:", data_add)
```

```python
data_inplace = data.add_(10) # 原地操作，直接修改原始张量的值
print("Data:", data)
print("Data + 10 (in-place):", data_inplace)
```

#### 5. 常用数学函数：

```python
# 幂运算
print(torch.pow(data,2))
# 平方根
print(torch.sqrt(data))
# 指数函数
print(torch.exp(data))
# 对数函数
print(torch.log(data))
# 三角函数
print(torch.sin(data))
print(torch.cos(data))
print(torch.tan(data))
```

#### 6. 比较运算：
1. 比较运算符：==、!=、>、<、>=、<=
2. 返回一个布尔类型的张量，表示每个元素的比较结果

```python
a = torch.tensor([1,2,3])
b = torch.tensor([3,2,1])
print(a == b) # [False, True, False]
print(a != b) # [True, False, True]
print(a > b)  # [False, False, True]
print(a < b)  # [True, False, False]
print(a >= b) # [False, True, True]
print(a <= b) # [True, True, False]
```

#### 7. 逻辑运算：
1. 逻辑运算符：&（与）、|（或）、~（非）
2. 返回一个布尔类型的张量，表示每个元素的逻辑运算结果

```python
a = torch.tensor([True, False, True])
b = torch.tensor([False, False, True])
print(a & b) # [False, False, True]
print(a | b) # [True, False, True]
print(~a)    # [False, True, False]
```

#### 8. 统计函数：

2. 这些函数可以对张量的元素进行统计计算，返回一个标量或者一个新的张量

```python
data = torch.tensor([[1,2,3],
                     [4,5,6]],dtype=torch.float32)
print("Sum:", data.sum()) # 所有元素的和
print("Mean:", data.mean().float()) # 所有元素的平均值
print("Std:", data.std()) # 所有元素的标准差
print("Var:", data.var()) # 所有元素的方差
print("Min:", data.min()) # 所有元素的最小值
print("Max:", data.max()) # 所有元素的最大值
```
