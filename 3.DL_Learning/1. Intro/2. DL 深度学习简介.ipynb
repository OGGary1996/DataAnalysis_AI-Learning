## 2. DL 深度学习简介

#### 1. 深度学习（Deep Learning）：
- 深度学习是一种基于人工神经网络的机器学习方法，通过多层神经网络模型来学习数据的高级特征表示
- 深度学习在图像识别、自然语言处理、语音识别等领域取得了显著的成果，成为当前人工智能领域的主流方法
- 深度学习的核心思想是通过多层神经网络模型来自动学习数据的特征表示，避免了传统机器学习方法中需要手工设计特征的过程

#### 2. 与传统机器学习的区别：
- 传统机器学习：需要手工设计特征，模型相对简单，适用于小规模数据集，容易过拟合
- 深度学习：自动学习特征表示，模型复杂，适用于大规模数据集，能够更好地泛化
- 传统机器学习：常用算法包括决策树、支持向量机、随机森林等
- 深度学习：常用模型包括全连接神经网络、卷积神经网络、循环神经网络、Transformer 等

#### 3. 深度学习的核心特点：

##### 3.1 端到端：
 - 原始数据直接输入模型，模型直接输出结果，中间不需要手工设计特征
 - 例如，图像分类任务中，原始图像直接输入卷积神经网络，模型直接输出分类结果

##### 3.2 层级特征学习：
- 深度学习通过多层神经网络模型来学习数据的层级特征表示，低层特征表示简单，高层特征表示复杂
- 例如，在图像识别任务中，低层特征可能是边缘和纹理，高层特征可能是物体的形状和类别

##### 3.3 强表达能力：
- 深度学习模型具有强大的表达能力，能够拟合复杂的函数关系，适用于处理非线性问题
- 例如，在自然语言处理任务中，深度学习模型能够捕捉文本中的语义关系和上下文信息，提升文本理解和生成的能力

##### 3.3 高度依赖反向传播算法：
- 深度学习模型的训练依赖于反向传播算法，通过计算损失函数的梯度来更新模型参数，优化模型性能
- 反向传播算法通过链式法则计算梯度，能够高效地训练深层神经网络模型，解决了传统机器学习中梯度消失和梯度爆炸的问题

#### 4. 常见的深度学习模型：
1. 全连接神经网络（ANN）：适用于处理结构化数据，如表格数据、时间序列数据
2. 卷积神经网络（CNN）：适用于处理图像数据，如图像分类、目标检测、图像分割等任务
  - CV：计算机视觉领域的核心模型，能够提取图像的局部特征和空间关系
3. 循环神经网络（RNN）：适用于处理序列数据，如文本数据、时间序列数据等
  - NLP：自然语言处理领域的核心模型，能够捕捉文本中的时间依赖关系和上下文信息
4. Transformer：适用于处理序列数据，特别是在自然语言处理领域，如机器翻译、文本生成等任务
5. BERT（Bidirectional Encoder Representations from Transformers）：基于 Transformer 的预训练语言模型，适用于自然语言理解任务，如文本分类、命名实体识别等
6. GPT（Generative Pre-trained Transformer）：基于 Transformer 的预训练语言模型，适用于自然语言生成任务，如文本生成、对话系统等
