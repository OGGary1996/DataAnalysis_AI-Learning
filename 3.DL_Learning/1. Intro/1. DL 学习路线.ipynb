## 1. DL 学习路线

#### 学习路线：
1. pytorch 张量运算，线性代数，自动求导，优化器，数据加载等基础知识
2. 神经网络基础：全连接层，卷积层，循环层，激活函数，损失函数等
3. pytorch 实战：实现经典网络（如 LeNet，AlexNet，VGG，ResNet 等），并在 MNIST，CIFAR-10 等数据集上进行训练和测试
4. ANN 实战：构建全连接神经网络，经典案例比如 MNIST 手写数字识别
5. CNN + CV 实战：构建卷积神经网络，经典案例比如 CIFAR-10 图像分类
6. RNN + NLP 实战：构建循环神经网络，经典案例比如 IMDB 电影评论情感分析
7. 进阶学习：Transformer，BERT，GPT 等最新模型的原理和实现

#### 1. 从标量到张量：
- 标量（scalar）：单个数值，如 3.14
- 向量（vector）：一维数组，如 [1, 2, 3]
- 矩阵（matrix）：二维数组，如 [[1, 2], [3, 4]]
- 张量（tensor）：多维数组，可以是任意维度，如 [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]
  - 深度学习所有的数据结构和运算全部围绕**张量**进行

#### 2. 框架对比：TensorFlow vs PyTorch
- TensorFlow：由 Google 开发，支持静态计算图，适合大规模分布式训练，社区活跃，生态丰富
- PyTorch：由 Facebook 开发，支持动态计算图，易于调试和开发，适合研究和快速原型开发，社区活跃，生态丰富
- 选择建议：初学者推荐 PyTorch，因为它更直观和易于学习；如果需要部署到生产环境或者进行大规模训练，可以考虑 TensorFlow

#### 3. ANN(Artificial Neural Network) 简介：
- 人工神经网络（ANN）是一种模拟生物神经网络的计算模型，由输入层、隐藏层和输出层组成
- 每层由多个神经元组成，神经元之间通过权重连接，输入数据通过权重进行加权求和，并通过激活函数进行非线性变换
- ANN 适用于处理结构化数据，如表格数据、时间序列数据

#### 4. CNN(Convolutional Neural Network) 简介：
- 卷积神经网络（CNN）是一种专门用于处理图像数据的神经网络，由卷积层、池化层和全连接层组成
- 卷积层通过卷积操作提取图像的局部特征，池化层通过下采样减少特征图的尺寸，全连接层进行分类或回归
- CNN 适用于处理图像数据，如图像分类、目标检测、图像分割等任务

#### 5. RNN(Recurrent Neural Network) 简介：
- 循环神经网络（RNN）是一种专门用于处理序列数据的神经网络，由循环层组成，能够捕捉序列中的时间依赖关系
- RNN 通过循环连接将前一个时间步的输出作为当前时间步的输入，适用于处理文本数据、时间序列数据等任务
- RNN 的变体包括 LSTM（长短期记忆网络）和 GRU（门控循环单元），它们通过引入门机制解决了传统 RNN 的梯度消失问题
