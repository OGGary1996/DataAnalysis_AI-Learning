## 2.2 损失函数的正规方程解 - 二元线性回归

#### 1. 手动计算损失函数的正规方程解<br>
此处损失函数使用SSE（Sum of Squared Errors，平方误差和）<br>
$$ SSE = \sum{(y_i - (w_1x_{i1} + w_2x_{i2} + b))^2} $$<br>
而由于二元线性回归中， x 作为矩阵存在， 所以损失函数作为欧式距离的平方，可以表示为：<br>
$$ SSE = (Y - XW)^T(Y - XW) $$<br>
针对W求导，并令其为0，可以得到正规方程解：<br>
$$ W = (X^TX)^{-1}X^TY $$<br>
其中，W为参数向量，包含w1, w2和b。<br>
针对b的求解，可以将X矩阵的第一列设为1，从而将b包含在W中。<br>
得到W后，可以直接从中提取出w1, w2和b。<br>

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# 使用简单的一组二元数据：
X = np.array([
    [1, 2],
    [2, 1],
    [3, 4],
    [4, 3]
], dtype=float)
y = np.array([3, 3, 7, 7], dtype=float)

# 添加偏置项
X_bias = np.c_[np.ones(X.shape[0]), X]

# 正规方程解计算W
w = np.linalg.inv(X_bias.T @ X_bias) @ X_bias.T @ y

w
```

## 2. 使用 sklearn 模型自动拟合计算<br>

```python
model = LinearRegression()
model.fit(X, y)
w_sklearn = np.concatenate(([model.intercept_], model.coef_))
w_sklearn
```
