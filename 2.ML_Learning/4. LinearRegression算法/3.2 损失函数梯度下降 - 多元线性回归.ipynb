## 3.2 损失函数梯度下降 - 多元线性回归 <br>
此处损失函数使用SSE（Sum of Squared Errors，平方误差和）<br>
$$ SSE = \sum{(y_i - (w_1x_{i1} + w_2x_{i2} + b))^2} $$<br>

#### 手写模拟梯度下降过程，计算多元线性回归模型的参数 w1, w2 和 b。

##### 1. 准备数据

```python
import numpy as np

# 特征矩阵 X（m 个样本，n 个特征）
X = np.array([
    [1, 2],
    [2, 1],
    [3, 4],
    [4, 3]
], dtype=float)

# 标签向量 y
y = np.array([3, 3, 7, 7], dtype=float)

m, n = X.shape
```

##### 2. 初始化参数，以 w1 = 0, w2 = 0, b = 0 开始， 学习率设置为0.01，迭代次数设置为20次

```python
# 权重向量（n 个参数）
w = np.zeros(n)

# 偏置
b = 0.0

# 学习率
alpha = 0.01

# 迭代次数
epochs = 20
```

##### 3. 代入迭代公式，计算 w1, w2 和 b 的更新

```python
for epoch in range(epochs):
    # 计算ŷ
    y_pred = X @ w + b  # 矩阵乘法

    # 计算误差
    error = y - y_pred

    # 计算损失 SSE
    loss = np.sum(error ** 2)

    # 梯度
    dw = -2 * X.T @ error
    db = -2 * np.sum(error)

    # 更新参数
    w = w - alpha * dw
    b = b - alpha * db

    print(f"Epoch {epoch+1}: w={w}, b={b:.4f}, Loss={loss:.4f}")
```
