## 2.1 损失函数的正规方程解 - 一元线性回归

#### 1. 手动计算损失函数的正规方程解<br>
此处损失函数使用SSE（Sum of Squared Errors，平方误差和）<br>
$$ SSE = \sum{(y_i - (wx_i + b))^2} $$<br>
令SSE对w和b的偏导数为0，解方程组可得w和b的解析解：<br>
$$ w= \frac{\sum_{i=1}^{n}(x_i-\bar{x})(y_i-\bar{y})}{\sum_{i=1}^{n}(x_i-\bar{x})^2} $$<br>
$$ b = \frac{(\sum y) - w(\sum x)}{n} $$<br>

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# 使用简单的一组数据：
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 5, 4, 5], dtype=float)

# 手动计算w 和 b
x_mean = x.mean()
y_mean = y.mean()

w = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean) ** 2)
b = y_mean - w * x_mean

print(f"Manual calculation - w: {w}, b: {b}")
```

#### 2. 使用 sklearn 模型自动拟合计算<br>
在使用 model.fit 方法后，模型会自动计算出最优的 w 和 b 参数。<br>
使用 model.coef_ 和 model.intercept_ 可以获取计算得到的 w 和 b 参数。<br>

```python
model = LinearRegression()
x_reshaped = x.reshape(-1, 1)  # sklearn 需要二维输入
model.fit(x_reshaped, y)
w_sklearn = model.coef_[0]
b_sklearn = model.intercept_
print(f"Sklearn calculation - w: {w_sklearn}, b: {b_sklearn}")
```
