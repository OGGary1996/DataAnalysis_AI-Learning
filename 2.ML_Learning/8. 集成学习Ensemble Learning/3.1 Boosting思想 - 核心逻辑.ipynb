## 3.1 Boosting思想 - 核心逻辑

#### 1. Boosting 是什么？
Boosting 是一种集成学习方法，通过将多个弱学习器（通常是简单的模型）组合成一个强学习器来提高预测性能。<br>
Boosting 的核心思想是通过迭代地训练弱学习器，每次训练时关注之前模型错误分类的样本，从而逐步提升整体模型的准确性。

#### 2. Boosting 要解决什么问题？
Boosting 主要解决：
- **减少偏差**：Boosting 通过关注错误分类的样本，逐步减少模型的偏差，从而提高整体性能。

#### 3. Boosting 的核心逻辑
Boosting 的核心逻辑可以总结为以下几个步骤：
1. **初始化权重**：为每个训练样本分配一个初始权重，通常是相等的。
2. **专注错误，迭代训练弱学习器**：在每一轮迭代中，训练一个弱学习器，并根据其性能调整样本的权重。错误分类的样本权重会增加，而正确分类的样本权重会减少。
3. **组合弱学习器**：将所有训练好的弱学习器组合成一个强学习器，通常通过加权投票或加权平均的方式进行预测。
4. **输出最终模型**：最终的模型是所有弱学习器的加权组合，能够更好地捕捉数据的复杂模式，从而提高预测性能。

#### 4. Boosting 的2大实现算法

##### 4.1 AdaBoost（Adaptive Boosting）
- **核心思想**：通过调整样本权重来专注于错误分类的样本。
- **算法步骤**：
  1. 初始化样本权重。
  2. 迭代训练弱学习器，并计算其错误率。
  3. 根据错误率调整样本权重，使得模型更加关注错误的样本
    - 分错的样本权证⬆
    - 分队的样本权重⬇
  4. 组合弱学习器，输出最终模型。

##### 4.2 Gradient Boosting （残差/梯度 Boosting）
- **核心思想**：通过优化损失函数来训练弱学习器。
- **算法步骤**：
  1. 初始化模型，通常是一个简单的常数模型；比如对于回归问题，可以使用训练数据的平均值作为初始模型；对于分类问题，可以使用训练数据中类别的比例作为初始模型。
  2. 第一次模型训练，得到当前模型的**残差（即预测值与真实值之间的差异）**，真实值 - 预测值 = 残差。
  3. 第二次模型训练，此时模型的目标不是直接尝试拟合真实值，而是尝试拟合结果为残差；
  4. 重复步骤2和3，直到此时总的残差越来越小，或者达到预设的迭代次数。

举例：
- 假设真实值为100，
- 第一次模型预测值为 y_pred_1 = 70，此时残差 r_1 = 100 - 70 = 30；
- 第二次模型的目标不再是直接尝试拟合100，而是尝试拟合上一次得到的残差 r_1 = 30;
  - 第二次的模型预测为 y_pred_2 = 25,此时新的残差 r_2 = 30 - 25 = 5;
- 第三次模型的目标是拟合新的残差 r_2 = 5;
  - 第三次的模型预测为 y_pred_3 = 4，此时新的残差 r_3 = 5 - 4 = 1;
- 最终的模型：
  - 最后的预测值为 = y_pred_1 + y_pred_2 + y_pred_3 = 70 + 25 + 4 = 99;
  - 最终的残差 r_final = 100 - 99= 1;

#### 5. Boosting 的核心流程：
1. **初始化模型**：选择一个简单的模型作为初始模型。
2. **迭代训练**：在每一轮迭代中：
  - AdaBoost：根据当前模型的错误分类情况调整样本权重，训练一个新的弱学习器。
  - Gradient Boosting：计算当前模型的残差，并训练一个新的弱学习器来拟合残差。
3. **更新模型**：将新训练的弱学习器加入到现有模型中，并更新模型的预测结果。
4. **重复迭代**：继续迭代，直到满足停止条件（如达到预设的迭代次数或残差足够小）。
5. **输出最终模型**：最终的模型是所有弱学习器的加权组合，能够更好地捕捉数据的复杂模式，从而提高预测性能。

#### 6. Boosting 为什么有效果？
Boosting 之所以有效，是因为它通过迭代地关注错误分类的样本，逐步减少模型的偏差，从而提高整体性能。<br>
每个弱学习器都在前一个模型的基础上进行改进，最终形成一个强大的模型，能够更好地捕捉数据的复杂模式。<br>
此外，Boosting 还具有一定的抗过拟合能力，因为它通过组合多个弱学习器来降低模型的方差，从而提高模型的泛化能力。

#### 7. Boosting 常用的弱学习器
Boosting 常用的弱学习器包括：
- **决策树**：特别是单层决策树（也称为决策树桩），因为它们简单且易于训练。
- **线性模型**：如线性回归或逻辑回归，适用于某些特定类型的数据。
- **其他简单模型**：如朴素贝叶斯、K近邻等，具体选择取决于数据的特性和问题的需求。
