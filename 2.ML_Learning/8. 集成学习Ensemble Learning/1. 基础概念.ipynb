## 1. Ensemble Learning 的基础概念

#### 1 什么是 Ensemble Learning
Ensemble Learning（集成学习）是一种机器学习方法，通过结合多个模型的预测结果来提高整体的性能和准确性。它的核心思想是通过集成多个弱学习器（即单个模型表现不佳的模型）来构建一个强学习器，从而获得更好的预测效果。

#### 2. Ensemble Learning 的核心思想
模型的合并思想： Voting（投票法）和 Averaging（平均法）和 Boosting 是3种常见的模型合并方法。
- Voting 适用于分类问题，通过多数投票来决定最终的预测结果；
- Averaging 适用于回归问题，通过计算多个模型预测值的平均值来得到最终的预测结果。
- Boosting 是一种迭代方法，通过逐步调整模型的权重来提高整体性能，常见的 Boosting 算法包括 AdaBoost 和 Gradient Boosting。

#### 3. Ensemble Learning的三大算法分类
- Bagging（Bootstrap Aggregating）：通过对训练数据进行有放回的抽样，生成多个不同的训练集，并在每个训练集上训练一个模型，最后通过投票或平均来得到最终的预测结果。常见的 Bagging 算法包括 Random Forest。
- Boosting：通过逐步调整模型的权重来提高整体性能，常见的 Boosting 算法包括 AdaBoost 和 Gradient Boosting。
- Stacking：通过训练一个元模型（meta-model）来结合多个基础模型的预测结果，从而得到最终的预测结果。元模型通常是一个简单的模型，如线性回归或逻辑回归。
