## 1. 决策树的基本原理和核心逻辑

#### 1. 决策树的基本原理
决策树是一种基于树形结构的分类和回归算法。它通过递归地将数据集划分成更小的子集，直到满足某个停止条件为止。<br>
决策树的基本原理是通过选择**最优的特征**来划分数据集，使得每个子集中的样本尽可能纯净（即属于同一类别）。 <br>
注意：
- 分类是通过特征来进行分类
- 每个节点代表一个特征，每个分支代表一个特征值的取值，而每个叶子节点代表一个类别或一个数值预测。

##### 1.1 根节点
根节点是决策树的起点，包含了整个数据集。它是决策树的第一个节点，负责对数据进行初始划分。根节点的选择对于决策树的性能至关重要，因为它决定了后续的划分方式和树的结构。<br>
根节点表示了整个数据集的特征和标签信息。它通过选择**一个特征**来划分数据集，使得每个子集中的样本尽可能纯净。<br>
根节点的选择通常基于某种指标，如信息增益、基尼指数或均方误差等。

##### 1.2 内部节点
内部节点是决策树中的非叶子节点，负责对数据进行进一步的划分。每个内部节点代表**一个特征**，通过对该特征的不同取值进行划分，将数据集分成更小的子集。<br>
内部节点的选择同样基于某种指标，如信息增益、基尼指数或均方误差等。通过选择最优的特征来划分数据集，内部节点帮助决策树逐步细化分类或回归的结果。<br>
内部节点的数量和深度会影响决策树的复杂度和性能。过多的内部节点可能导致过拟合，而过少的内部节点可能导致欠拟合。

##### 1.3 叶子节点
叶子节点是决策树中的终端节点，代表了最终的分类结果或回归预测。每个叶子节点对应**一个类别标签（对于分类问题）或一个数值预测（对于回归问题）**。<br>
注意：
- 叶子阶段不再表示特征，而是表示分类结果或者数值预测 <br>
- 叶子节点的值是通过对数据集进行划分后得到的子集中的样本进行统计或计算得出的。对于分类问题，叶子节点的值通常是子集中样本所属的多数类别；对于回归问题，叶子节点的值通常是子集中样本的平均值或中位数。<br>
叶子节点的数量和分布会影响决策树的性能。过多的叶子节点可能导致过拟合，而过少的叶子节点可能导致欠拟合。因此，在构建决策树时，需要合理地控制叶子节点的数量和深度，以获得更好的泛化能力。

##### 1.4 路径
路径是从根节点到叶子节点的一条连续的分支，代表了决策树中的一个决策过程。每条路径由一系列的内部节点和一个叶子节点组成，描述了从输入特征到最终分类或回归结果的决策流程。<br>
路径的长度和复杂度会影响决策树的性能。过长的路径可能导致过拟合，而过短的路径可能导致欠拟合。因此，在构建决策树时，需要合理地控制路径的长度和复杂度，以获得更好的泛化能力。<br>
路径的选择通常基于某种指标，如信息增益、基尼指数或均方误差等。通过选择最优的路径来划分数据集，决策树能够逐步细化分类或回归的结果，从而提高模型的性能和准确率。

#### 2. 决策树的核心逻辑
决策树的核心逻辑是通过递归地划分数据集来构建树形结构。决策树的构建过程可以分为以下几个步骤：
1. **选择最优特征**：在每个节点上，选择**一个特征**来划分数据集。选择的标准通常是基于某种指标，如**信息增益、基尼指数或均方误差**等。通过选择最优的特征来划分数据集，使得每个子集中的样本尽可能纯净。
2. **划分数据集**：根据**选择的特征**，将数据集划分成更小的子集。**每个子集对应一个特征值的取值**。通过划分数据集，决策树能够逐步细化分类或回归的结果。
3. **递归构建子树**：对于每个子集，重复上述步骤，选择最优特征并划分数据集，直到满足某个停止条件为止。停止条件可以是达到最大树深度、叶子节点的样本数量小于某个阈值，或者所有样本属于同一类别等。
4. **生成叶子节点**：当满足停止条件时，生成叶子节点。对于分类问题，叶子节点的值通常是子集中样本所属的多数类别；对于回归问题，叶子节点的值通常是子集中样本的平均值或中位数。通过生成叶子节点，决策树能够提供最终的分类或回归结果。
通过以上步骤，决策树能够逐步构建出一个树形结构，从而实现对数据的分类或回归。决策树的核心逻辑在于通过选择最优特征来划分数据集，使得每个子集中的样本尽可能纯净，从而提高模型的性能和准确率。

#### 3. 分裂的方式 & 得到的结果
分裂的方式（离散特征和连续特征）：
- 离散特征：对于离散特征，决策树会根据特征的不同取值将数据集划分成多个子集。每个子集对应一个特征值的取值。
  - 比如：如果一个特征 Weather 有三个不同的取值{Sunny, Overcast, Rain}，那么决策树会将数据集划分成三个子集，每个子集包含对应特征值的样本。
    - S_sunny：Weather=Sunny 的所有样本
    - S_overcast：Weather=Overcast 的所有样本
    - S_rain：Weather=Rain 的所有样本
- 连续特征：对于连续特征，决策树会选择一个阈值将数据集划分成两个子集。一个子集包含特征值小于或等于阈值的样本，另一个子集包含特征值大于阈值的样本。
  - 例如，如果一个特征是年龄，决策树可能会选择一个年龄阈值（如50岁）来划分数据集，将年龄小于或等于30岁的样本划分到一个子集，年龄大于30岁的样本划分到另一个子集。
    - S_age_leq_30：年龄小于或等于30岁的所有样本
    - S_age_gt_30：年龄大于30岁的所有样本

不同的算法的分裂方式不同：
- ID3 & C4.5 算法：多分叉，即对于离散特征，决策树会根据特征的不同取值将数据集划分成多个子集，每个子集对应一个特征值的取值。
- CART算法：二分叉，即对于连续特征，决策树会选择一个阈值将数据集划分成两个子集，一个子集包含特征值小于或等于阈值的样本，另一个子集包含特征值大于阈值的样本；即使是离散特征，CART算法也会将其转换为二元特征进行划分。
