## 1. Naive Bayes核心概念

#### 1. 核心目的：
1. 输入：一组样本的特征值（属性值）和对应的类别标签。
2. 输出：预测类别标签。
3. 核心任务：比较不同类别的后验概率，选择具有最高后验概率的类别作为预测结果。
y = argmax P(C|X) = argmax P(X|C) * P(C)

#### 2. 贝叶斯公式，为什么能做分类：
1. 贝叶斯公式：P(C|X) = P(X|C) * P(C) / P(X)
2. 分类任务中，我们关注的是比较不同类别的后验概率 P(C|X)，而 P(X) 是一个常数，可以忽略。因此，我们可以简化为：P(C|X) ∝ P(X|C) * P(C)
3. 通过计算 P(X|C) 和 P(C)，我们可以比较不同类别的后验概率，选择具有最高后验概率的类别作为预测结果。

#### 3. 朴素假设：
1. 朴素假设：假设特征之间是条件独立的，即 P(X|C) = P(x1|C) * P(x2|C) * ... * P(xn|C)
2. 这种假设简化了计算，使得我们可以独立地计算每个特征的条件概率 P(xi|C)，然后将它们相乘得到 P(X|C)。
3. 尽管这种假设在现实中可能不完全成立，但在许多实际应用中，朴素贝叶斯分类器仍然表现出良好的性能。

#### 4. 最终决策：贝叶斯 + 朴素假设
1. 最终决策：y = argmax P(C|X) = argmax P(X|C) * P(C) = argmax P(x1|C) * P(x2|C) * ... * P(xn|C) * P(C)
2. 通过计算每个类别的后验概率，我们可以选择具有最高后验概率的类别作为预测结果。
3. 这种方法结合了贝叶斯定理和朴素假设，使得分类任务变得简单且高效，尤其适用于高维数据和大规模数据集。

#### 5. 平滑：Laplace Smoothing
1. 平滑的目的：当某个类别在训练数据中没有出现某个特征值时，P(xi|C) = 0，这会导致整个 P(X|C) = 0，从而使得后验概率 P(C|X) = 0，无法进行有效的分类。
2. Laplace Smoothing：通过在每个特征值的计数中添加一个小的常数（通常是1），来避免零概率问题。公式如下：
   P(xi|C) = (count(xi,C) + 1) / (count(C) + k)，其中 k 是特征值的总数。
3. 这种方法确保了即使某个特征值在训练数据中没有出现，也不会导致整个概率为零，从而提高了模型的鲁棒性。

#### 6. 避免连乘下溢出：对数变换
1. 连乘下溢出问题：当特征数量较多时，P(X|C) = P(x1|C) * P(x2|C) * ... * P(xn|C) 可能会变得非常小，导致数值下溢出。
2. 对数变换：通过对概率取对数，将连乘转化为连加，避免下溢出问题。公式如下：
   log P(C|X) = log P(X|C) + log P(C) = log P(x1|C) + log P(x2|C) + ... + log P(xn|C) + log P(C)
3. 通过使用对数变换，我们可以在计算过程中保持数值的稳定性，避免下溢出问题，同时仍然能够正确地比较不同类别的后验概率。

#### 7. 三种常见的朴素贝叶斯分类器：
1. Gaussian Naive Bayes（高斯朴素贝叶斯）：适用于连续特征，假设特征值服从高斯分布。
2. Multinomial Naive Bayes（多项式朴素贝叶斯）：适用于离散特征，常用于文本分类，假设特征值服从多项式分布。
3. Bernoulli Naive Bayes（伯努利朴素贝叶斯）：适用于二元特征，常用于文本分类，假设特征值服从伯努利分布。
