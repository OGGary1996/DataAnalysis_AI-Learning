## 3. SVM 核函数 Kernel Trick

#### 1. 核函数的概念<br>
在实际应用中，数据往往是非线性可分的。为了处理这种情况，我们引入核函数（Kernel Function）。<br>
核函数是一种将数据映射到高维空间的技术，使得在高维空间中数据变得线性可分。<br>
通过使用核函数，我们可以在不显式计算高维映射的情况下，计算样本点之间的内积。<br>
常用的核函数包括线性核、多项式核和径向基函数（RBF）核。<br>
- 通过引入核函数，SVM 能够处理复杂的非线性分类问题，扩展了其应用范围。

#### 2. 核函数的分类<br>
常用的核函数包括以下几种类型：<br>
- 线性核（Linear Kernel）：适用于线性可分的数据集，计算样本点之间的内积。<br>
- 多项式核（Polynomial Kernel）：适用于非线性数据，通过多项式映射将数据映射到高维空间。<br>
- 径向基函数核（Radial Basis Function, RBF Kernel）：适用于复杂的非线性数据，通过高斯函数映射数据。<br>
